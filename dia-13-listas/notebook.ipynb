## 🧠 Conceito em 2 minutos

Uma **lista** é uma fila numerada de itens: `["BH", "Betim", "Contagem"]`. A numeração começa em
**0**, e `lista[-1]` é sempre o último.

| Quero… | Use | Muda a lista? |
|---|---|---|
| pôr no fim | `lista.append(x)` | sim |
| pôr na posição i | `lista.insert(i, x)` | sim |
| tirar da posição i | `lista.pop(i)` (devolve o item) | sim |
| tirar pelo valor | `lista.remove(x)` | sim |
| ordenar | `lista.sort()` · `sorted(lista)` | sim · **não** (devolve outra) |
| copiar | `lista.copy()` | não |
| tirar repetidos com ordem | `list(dict.fromkeys(lista))` | não |
| contar | `Counter(lista).most_common(3)` | não |

**Ordenar com acento:** `sorted()` compara os códigos Unicode das letras. Todas as maiúsculas vêm
antes das minúsculas, e as acentuadas vêm depois do "z". A solução é dar ao `sorted` uma **chave**
(`key=`): uma função que transforma cada item numa versão "comparável" (sem acento, tudo minúsculo).
O item original não muda; só a comparação.

**Conjunto (`set`)** é uma "lista sem repetidos e sem ordem". Serve para perguntar "o que tem nos
dois?" (`a & b`), "o que só tem em A?" (`a - b`) e "o que tem em algum?" (`a | b`).

### `br/01_capitais.py` — As operações do original (copy, sort, insert, append, pop) com capitais brasileiras.

````python
%cd -q {PASTA_NB}
"""
As operações do original (copy, sort, insert, append, pop) com capitais brasileiras.

Carta 34 (rank A) — Jutsu da Lista Ordenada

O que você aprende:
  • a criar, copiar, ordenar, inserir e remover itens de uma lista;
  • a diferença entre copiar (copy) e só dar outro nome (=);
  • a diferença entre sort() (muda a lista) e sorted() (devolve uma nova);
  • a usar índices, inclusive negativos, e o operador "in".

Como rodar (dentro da pasta dia-13-listas):
    python br/01_capitais.py
"""

# ── 1. Criar a lista ──────────────────────────────────────────────────────
# Uma lista fica entre colchetes; cada item, entre aspas, separado por vírgula.
capitais = ["Belo Horizonte", "São Paulo", "Rio de Janeiro", "Vitória"]
print("1) Sudeste:", capitais)
print(f"   São {len(capitais)} capitais. A primeira é {capitais[0]}; a última é {capitais[-1]}.")
#   capitais[0]  → primeira (a contagem começa em ZERO)
#   capitais[-1] → última (negativo conta de trás para frente)

# ── 2. Copiar: copy() x "=" ───────────────────────────────────────────────
copia = capitais.copy()        # lista NOVA, independente
apelido = capitais             # NÃO é cópia: é outro nome para a MESMA lista
apelido.append("Brasília")     # mexemos pelo "apelido"...
print("\n2) Depois de apelido.append('Brasília'):")
print("   capitais:", capitais)  # ...e a original mudou junto!
print("   copia:   ", copia)     # a cópia de verdade ficou intacta
capitais.pop()                 # desfaz: tira o último item (Brasília)

# ── 3. Acrescentar no fim: append() ───────────────────────────────────────
capitais.append("Curitiba")
print("\n3) append('Curitiba'):", capitais)

# ── 4. Inserir numa posição: insert(posição, item) ────────────────────────
capitais.insert(1, "Salvador")        # entra na posição 1 (a segunda); o resto anda uma casa
print("4) insert(1, 'Salvador'):", capitais)

# ── 5. Remover pela posição: pop(posição) ─────────────────────────────────
removida = capitais.pop(2)            # pop devolve o item removido: dá para guardar
print(f"5) pop(2) removeu '{removida}':", capitais)

# ── 5b. Remover pelo valor: remove(item) ──────────────────────────────────
if "Curitiba" in capitais:            # "in" pergunta se o item está na lista
    capitais.remove("Curitiba")       # remove a PRIMEIRA ocorrência; erro se não existir
print("   remove('Curitiba'):", capitais)

# ── 6. Ordenar: sort() e sorted() ─────────────────────────────────────────
nova = sorted(capitais)               # devolve uma lista nova ordenada
print("\n6) sorted(capitais):", nova)
print("   capitais continua:", capitais)
capitais.sort(reverse=True)           # ordena a própria lista, de Z para A
print("   capitais.sort(reverse=True):", capitais)

# ── 7. Imprimir bonito: * e sep ───────────────────────────────────────────
print("\n7) Com separador:")
print(*sorted(capitais), sep=" · ")   # o * "espalha" a lista em vários argumentos

# ── 8. Bônus: listas de listas (capital, UF) ──────────────────────────────
pares = [["Belo Horizonte", "MG"], ["Vitória", "ES"], ["Salvador", "BA"], ["São Paulo", "SP"]]
pares.sort(key=lambda par: par[1])    # key diz "ordene olhando para..." (aqui, a UF)
print("\n8) Ordenado pela UF:")
for capital, uf in pares:             # "desempacota" cada par em duas variáveis
    print(f"   {uf} → {capital}")
````

### `br/02_ordenar_com_acento.py` — Por que "Águas Lindas" vai parar depois de "Zé Doca"? Ordenação com acentos.

````python
%cd -q {PASTA_NB}
"""
Por que "Águas Lindas" vai parar depois de "Zé Doca"? Ordenação com acentos.

Carta 34 (rank A) — Jutsu da Lista Ordenada (parte 2)

O que você aprende:
  • que sorted() compara letras pelo código Unicode: maiúsculas antes de minúsculas,
    e letras acentuadas depois do "z";
  • a criar uma "chave de ordenação" sem acento e sem diferença de caixa
    (unicodedata.normalize + casefold);
  • a desempatar palavras iguais sem acento ("Pará" x "para");
  • que locale.strxfrm resolve, mas depende do sistema ter o idioma pt_BR instalado.

Como rodar (dentro da pasta dia-13-listas):
    python br/02_ordenar_com_acento.py
"""

import locale
import unicodedata

CIDADES = ["Zé Doca", "Águas Lindas de Goiás", "Belo Horizonte", "Óbidos",
           "abaetetuba", "Itaúna", "Ibirité", "Ouro Preto", "Uberlândia", "Ubá"]

# ── 1. O problema ─────────────────────────────────────────────────────────
print("❌ sorted() comum:")
for c in sorted(CIDADES):
    print(f"   {c}")
print("\n   Por quê? Cada letra tem um número (ord). Veja:")
for letra in ["B", "Z", "a", "z", "Á", "Ó"]:
    print(f"   ord('{letra}') = {ord(letra)}")
print("   Maiúsculas (65–90) < minúsculas (97–122) < acentuadas (192+).")


# ── 2. A solução: uma chave "sem acento e sem caixa" ──────────────────────
def chave_pt(texto):
    """
    'Águas' → 'aguas'. Passo a passo:
      1. normalize("NFKD") separa a letra do acento: 'Á' vira 'A' + '´' (acento solto);
      2. o filtro joga fora os acentos soltos (combining);
      3. casefold() deixa tudo minúsculo (é um lower() mais caprichado: 'ß' → 'ss').
    Devolve uma TUPLA: (sem acento, original). A segunda parte desempata
    palavras que ficam iguais sem acento, de forma sempre previsível.
    """
    separado = unicodedata.normalize("NFKD", texto)
    sem_acento = "".join(c for c in separado if not unicodedata.combining(c))
    return (sem_acento.casefold(), texto)


print("\n✅ sorted(key=chave_pt):")
for c in sorted(CIDADES, key=chave_pt):
    print(f"   {c:<24} chave → {chave_pt(c)[0]}")

# ── 3. Desempate ──────────────────────────────────────────────────────────
print("\n🔁 Desempate com palavras iguais sem acento:")
print("  ", sorted(["para", "Pará", "pára"], key=chave_pt))

# ── 4. A alternativa do sistema: locale.strxfrm ───────────────────────────
# Se o sistema tiver o idioma pt_BR, o próprio sistema sabe ordenar do jeito brasileiro.
# Em muitos servidores (e no Colab) ele não vem instalado: por isso a chave acima é mais garantida.
print("\n🌐 Tentando locale pt_BR:")
for nome in ("pt_BR.UTF-8", "pt_BR.utf8", "Portuguese_Brazil.1252"):   # Linux/macOS e Windows
    try:
        locale.setlocale(locale.LC_COLLATE, nome)
        print(f"   ✅ locale '{nome}' disponível:")
        print("  ", sorted(CIDADES, key=locale.strxfrm))
        break
    except locale.Error:
        continue
else:
    # O "else" de um for roda quando o laço termina SEM break.
    print("   ⚠️ pt_BR não está instalado neste sistema; fique com a chave_pt.")
````

### `br/03_cidades_das_vitimas.py` — O mapa das vítimas: tirar repetidos, fazer ranking e comparar outubro x novembro.

````python
%cd -q {PASTA_NB}
"""
O mapa das vítimas: tirar repetidos, fazer ranking e comparar outubro x novembro.

Carta 34 (rank A) — Jutsu da Lista Ordenada (aplicação no caso)

O que você aprende:
  • a montar uma lista a partir de uma coluna do CSV;
  • dict.fromkeys(lista): tira repetidos MANTENDO a ordem em que apareceram;
  • set(lista): tira repetidos, mas sem ordem (e permite comparar grupos);
  • collections.Counter: conta quantas vezes cada item aparece (ranking);
  • operações de conjunto: & (nos dois), - (só em um), | (em algum).

Este script é 100% offline: lê caso/dados/denuncias.csv (fictício).

Como rodar (dentro da pasta dia-13-listas):
    python br/03_cidades_das_vitimas.py
"""

import csv
import unicodedata
from collections import Counter
from pathlib import Path

TOP = 5  #@param {type:"integer"}

PASTA_DIA = Path(__file__).resolve().parent.parent if "__file__" in globals() else Path.cwd()
PASTA_CASO = PASTA_DIA.parent / "caso" / "dados"

# Nomes oficiais (grafia do IBGE) das cidades que aparecem no caso.
# No Dia 7 você gerou essa lista automaticamente com a API do IBGE.
OFICIAIS = ["Belo Horizonte", "Betim", "Contagem", "Divinópolis", "Governador Valadares",
            "Ipatinga", "Juiz de Fora", "Poços de Caldas", "Sete Lagoas", "Uberaba", "Uberlândia"]
# Erros de digitação conhecidos (no Dia 7, o difflib descobriu isso sozinho).
CORRECOES = {"belo horizone": "belo horizonte"}


def sem_acento(texto):
    """'Poços de Caldas' → 'pocos de caldas' (mesma ideia do script 02)."""
    separado = unicodedata.normalize("NFKD", texto.strip())
    return "".join(c for c in separado if not unicodedata.combining(c)).casefold()


# Dicionário "chave sem acento → nome oficial", montado com uma compreensão de dicionário.
OFICIAL_POR_CHAVE = {sem_acento(nome): nome for nome in OFICIAIS}


def padronizar(cidade):
    chave = sem_acento(cidade)
    chave = CORRECOES.get(chave, chave)
    return OFICIAL_POR_CHAVE.get(chave, cidade)    # se não reconhecer, devolve como veio


# ── 1. Ler a coluna "cidade" (e o mês) ────────────────────────────────────
with open(PASTA_CASO / "denuncias.csv", encoding="utf-8") as arquivo:
    linhas = list(csv.DictReader(arquivo, delimiter=";"))

cidades_brutas = [linha["cidade"] for linha in linhas]
print(f"📨 {len(cidades_brutas)} denúncias.")

# ── 2. Repetidos: set x dict.fromkeys ─────────────────────────────────────
print(f"\n🔤 Grafias distintas (como as vítimas digitaram): {len(set(cidades_brutas))}")
for c in dict.fromkeys(cidades_brutas):          # ordem de aparição preservada
    print(f"   • {c}")

cidades = [padronizar(c) for c in cidades_brutas]   # compreensão de lista: aplica a função em todos
unicas = list(dict.fromkeys(cidades))
print(f"\n🏙️  Depois de padronizar: {len(unicas)} cidades (na ordem em que surgiram):")
print("   " + " · ".join(unicas))

# ── 3. Ranking com Counter ────────────────────────────────────────────────
contagem = Counter(cidades)
print(f"\n🏆 Top {TOP} cidades com mais vítimas:")
for posicao, (cidade, qtd) in enumerate(contagem.most_common(TOP), start=1):
    print(f"   {posicao}º {cidade:<22} {'█' * qtd} {qtd}")

# ── 4. Outubro x novembro com conjuntos ───────────────────────────────────
# A data vem como "2025-11-30": os 7 primeiros caracteres são "2025-11".
outubro = {padronizar(l["cidade"]) for l in linhas if l["data"].startswith("2025-10")}
novembro = {padronizar(l["cidade"]) for l in linhas if l["data"].startswith("2025-11")}

print("\n📅 Outubro x novembro:")
print(f"   Nos dois meses (&):      {sorted(outubro & novembro)}")
print(f"   Só em outubro (-):       {sorted(outubro - novembro)}")
print(f"   Só em novembro (-):      {sorted(novembro - outubro)}")
print(f"   Em algum mês (|):        {len(outubro | novembro)} cidades")

print("\n🧠 Leitura de analista: as cidades que só aparecem em novembro mostram o golpe")
print("   se espalhando pelo estado perto da Black Friday.")
````

## ⚖️ Ética, LGPD e OPSEC

- **Duplicata infla o número.** Antes de dizer "40 vítimas", confirme que não há a mesma denúncia
  contada duas vezes (mesmo CPF, mesma data, mesmo valor). Número errado em relatório vira manchete
  errada.
- **Agregar protege.** Para a imprensa ou para um painel, mostre "8 vítimas em Belo Horizonte", não a
  lista de nomes. Cidade + data + valor de uma cidade pequena pode reidentificar a pessoa (LGPD:
  minimização e anonimização).
- **Padronizar não é corrigir a fonte.** Guarde o valor original ao lado do padronizado: se a
  padronização errar, o original prova o que a vítima informou.
- **Cuidado com `=`**: "copiar" uma lista de evidências com `=` e depois apagar itens da "cópia" apaga
  da original também. Em evidência, use sempre `copy()` (ou trabalhe numa cópia do arquivo).

# 🏆 Desafio S — Dia 13: Fila de diligências

**Carta 55 · Rank S**

## A missão

A Delegada Rita só tem efetivo para pedir os dados de **uma conta por vez** ao banco. Por qual chave
PIX começar? A Iara quer uma "fila de diligências" pronta para ir no ofício.

Escreva um script que, a partir de `caso/dados/denuncias.csv`:

1. **Agrupe** as denúncias por `chave_pix_destino` (uma lista de denúncias por chave).
2. **Ordene** cada lista pela data (e pelo `id`, para desempatar).
3. Para cada chave, calcule: número de vítimas, **total** com `Decimal`, primeira e última denúncia
   (`lista[0]` e `lista[-1]`), a cidade com mais vítimas (`Counter`) e as cidades em ordem alfabética
   **correta** (com acento e sem repetidos).
4. **Ordene a fila** pelo total (maior prejuízo primeiro); empate → mais vítimas primeiro.
5. Imprima e grave `saidas/fila_diligencias.txt` com os valores em formato brasileiro (`R$ 21.242,10`).

## Critérios de rank S

- [ ] Só biblioteca padrão (`csv`, `decimal`, `collections`, `unicodedata`)
- [ ] Cidades padronizadas antes de contar ("belo horizonte" e "Belo Horizone" são a mesma)
- [ ] `sort(key=...)` com **tupla** para desempatar
- [ ] O total geral bate com a soma de todas as denúncias

## Dica

`collections.defaultdict(list)` cria a lista vazia sozinho na primeira vez que a chave aparece:
`por_chave[chave].append(denuncia)`.

````python
%cd -q {PASTA_NB}
# ✍️ Escreva sua solução aqui.
````

````python
#@title 🔒 Gabarito (dê dois cliques aqui para ver o código) { display-mode: "form" }
%cd -q {PASTA_NB}
"""
Gabarito do Desafio S do Dia 13 — Fila de diligências por chave PIX.

Para cada chave PIX do caso, monta a lista de vítimas em ordem de data, calcula
total (Decimal), primeira e última denúncia, a cidade mais frequente e as cidades
em ordem alfabética correta (com acento). Depois ordena as chaves pelo valor total:
a de maior prejuízo vira a primeira diligência. Grava saidas/fila_diligencias.txt.
100% offline: só caso/dados (fictício).
"""

import csv
import unicodedata
from collections import Counter, defaultdict
from decimal import Decimal
from pathlib import Path

PASTA_DIA = Path(__file__).resolve().parent if "__file__" in globals() else Path.cwd()
PASTA_CASO = PASTA_DIA.parent / "caso" / "dados"

OFICIAIS = ["Belo Horizonte", "Betim", "Contagem", "Divinópolis", "Governador Valadares",
            "Ipatinga", "Juiz de Fora", "Poços de Caldas", "Sete Lagoas", "Uberaba", "Uberlândia"]


def sem_acento(texto):
    separado = unicodedata.normalize("NFKD", texto.strip())
    return "".join(c for c in separado if not unicodedata.combining(c)).casefold()


OFICIAL = {sem_acento(n): n for n in OFICIAIS}
OFICIAL["belo horizone"] = "Belo Horizonte"     # erro de digitação conhecido


def brl(valor):
    """Decimal('1999.9') → 'R$ 1.999,90' (troca de separadores via um caractere temporário)."""
    return "R$ " + f"{valor:,.2f}".replace(",", "_").replace(".", ",").replace("_", ".")


# ── 1. Agrupar as denúncias por chave ─────────────────────────────────────
por_chave = defaultdict(list)                  # chave → lista de denúncias (dicionários)
with open(PASTA_CASO / "denuncias.csv", encoding="utf-8") as f:
    for linha in csv.DictReader(f, delimiter=";"):
        linha["cidade"] = OFICIAL.get(sem_acento(linha["cidade"]), linha["cidade"])
        linha["valor"] = Decimal(linha["valor"])
        por_chave[linha["chave_pix_destino"]].append(linha)

# ── 2. Resumo de cada chave ───────────────────────────────────────────────
fila = []
for chave, lista in por_chave.items():
    lista.sort(key=lambda d: (d["data"], d["id"]))   # data ISO ordena certo como texto
    cidades = [d["cidade"] for d in lista]
    fila.append({
        "chave": chave,
        "vitimas": len(lista),
        "total": sum((d["valor"] for d in lista), Decimal("0")),
        "primeira": lista[0]["data"],                 # [0] = primeiro da lista ordenada
        "ultima": lista[-1]["data"],                  # [-1] = último
        "cidade_top": Counter(cidades).most_common(1)[0],
        "cidades": sorted(set(cidades), key=lambda c: (sem_acento(c), c)),
        "ids": [d["id"] for d in lista],
    })

# ── 3. Ordenar a fila: maior prejuízo primeiro (desempate: mais vítimas) ──
fila.sort(key=lambda item: (item["total"], item["vitimas"]), reverse=True)

linhas_relatorio = ["FILA DE DILIGÊNCIAS POR CHAVE PIX (maior prejuízo primeiro)", ""]
for posicao, item in enumerate(fila, start=1):
    bloco = [
        f"{posicao}ª diligência · chave {item['chave']}",
        f"   {item['vitimas']} vítimas · {brl(item['total'])} · de {item['primeira']} a {item['ultima']}",
        f"   cidade com mais vítimas: {item['cidade_top'][0]} ({item['cidade_top'][1]})",
        f"   cidades: {', '.join(item['cidades'])}",
        f"   denúncias: {', '.join(item['ids'][:6])}{' …' if len(item['ids']) > 6 else ''}",
        "",
    ]
    linhas_relatorio += bloco
total_geral = sum((i["total"] for i in fila), Decimal("0"))
linhas_relatorio.append(f"TOTAL GERAL: {brl(total_geral)} em {sum(i['vitimas'] for i in fila)} denúncias")

print("\n".join(linhas_relatorio))
saidas = PASTA_DIA / "saidas"
saidas.mkdir(exist_ok=True)
(saidas / "fila_diligencias.txt").write_text("\n".join(linhas_relatorio) + "\n", encoding="utf-8")
print("\n💾 saidas/fila_diligencias.txt gravado.")
````
