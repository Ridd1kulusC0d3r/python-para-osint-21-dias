## 🧠 Conceito em 2 minutos

Um **script** é um arquivo de texto terminado em `.py`. Quando você roda `python arquivo.py`, o
**interpretador** lê o arquivo de cima para baixo e executa cada linha.

entre aspas (simples ou duplas, tanto faz). Para misturar texto e valores, use uma **f-string**:

```python
nome = "Iara"

```

Para registros de investigação, use `datetime.now(ZoneInfo("America/Sao_Paulo"))`: a hora sai certa e
o fuso fica gravado junto (`2026-10-12T09:03:11-03:00`).

### `br/01_ola_nfa.py` — Olá, NFA: o seu primeiro script com nome, emoji e a hora certa de Brasília.

````python
%cd -q {PASTA_NB}
"""
Olá, NFA: o seu primeiro script com nome, emoji e a hora certa de Brasília.

Carta 22 (rank B) — Jutsu do Primeiro Alerta

O que você aprende:
  • print() com vários pedaços, com sep= e com end=;
  • f-strings: colocar o valor de uma variável dentro do texto com {chaves};
  • que emoji também é texto (é só mais um caractere Unicode);
  • a pegar a data e a hora no fuso America/Sao_Paulo, com plano B se o fuso faltar.

Como rodar (dentro da pasta dia-01-primeiro-script):
    python br/01_ola_nfa.py
"""

from datetime import datetime, timedelta, timezone  # tudo o que precisamos para datas e fusos

# ── Parâmetros (no Colab viram um formulário) ─────────────────────────────
# Troque o valor entre aspas e rode de novo. É assim que se "personaliza" um script.
MEU_NOME = "Analista"  #@param {type:"string"}
UNIDADE = "Núcleo de Fontes Abertas (NFA)"  #@param {type:"string"}

# ── 1. O print mais simples ───────────────────────────────────────────────
# Igual ao do cipher387: um texto entre aspas dentro de print().
print("Bem-vindo(a) ao curso de 21 dias de Python para OSINT!")

# ── 2. Vários pedaços no mesmo print ──────────────────────────────────────
# Separados por vírgula, o print junta os pedaços com um espaço no meio.
print("Unidade:", UNIDADE)
# sep= troca o separador; aqui usamos " · " em vez do espaço.
print("Belo Horizonte", "MG", "Brasil", sep=" · ")

# ── 3. f-string: o jeito moderno de montar textos ─────────────────────────
# O "f" antes das aspas avisa: "procure {variáveis} aqui dentro e troque pelo valor".
print(f"👋 Olá, {MEU_NOME}! Seu crachá do {UNIDADE} está ativo.")

# Dentro das chaves pode ir uma conta ou uma função, não só o nome da variável.
print(f"🔠 Seu nome tem {len(MEU_NOME)} letras e, em maiúsculas, fica {MEU_NOME.upper()}.")

# ── 4. Emojis são texto ───────────────────────────────────────────────────
# Cada emoji é um caractere como qualquer letra. Dá para repetir texto com *.
print("🚨" * 3, "ALERTA DE GOLPE", "🚨" * 3)
# end= muda o que o print coloca no final (o padrão é pular linha, "\n").
print("Carregando", end="")
print("...", end="")
print(" pronto! ✅")

# ── 5. Data e hora de Brasília ────────────────────────────────────────────
# Em investigação, horário SEMPRE com fuso. "14h" sem fuso não serve como registro.
# zoneinfo (Python 3.9+) conhece os fusos do mundo todo. No Windows, pode faltar o
# banco de fusos (pacote tzdata): nesse caso usamos UTC-3 fixo, que vale para Brasília
# desde 2019, quando o horário de verão foi extinto.
try:
    from zoneinfo import ZoneInfo
    FUSO = ZoneInfo("America/Sao_Paulo")
except Exception:
    FUSO = timezone(timedelta(hours=-3))

agora = datetime.now(FUSO)

# strftime transforma a data em texto, no formato que quisermos:
#   %d = dia, %m = mês, %Y = ano com 4 dígitos, %H:%M:%S = hora:minuto:segundo
print(f"\n🕒 Agora em Brasília: {agora:%d/%m/%Y %H:%M:%S}")
# isoformat() é o formato internacional (bom para registros e planilhas).
print(f"🗂️  Formato ISO 8601:  {agora.isoformat(timespec='seconds')}")

# Os dias da semana em português, sem depender da configuração do computador.
DIAS = ["segunda-feira", "terça-feira", "quarta-feira", "quinta-feira",
        "sexta-feira", "sábado", "domingo"]
# weekday() devolve 0 para segunda, 1 para terça... e 6 para domingo.
print(f"📅 Hoje é {DIAS[agora.weekday()]}.")

print("\n✅ Seu primeiro script rodou. Carta 22 conquistada!")
````

### `br/02_briefing_caso.py` — Briefing do caso: o primeiro relatório da Operação Vitrine Fantasma na tela.

````python
%cd -q {PASTA_NB}
"""
Briefing do caso: o primeiro relatório da Operação Vitrine Fantasma na tela.

Carta 22 (rank B) — Jutsu do Primeiro Alerta (parte 2)

O que você aprende:
  • a guardar os números do caso em variáveis (texto, inteiro, decimal);
  • a montar um texto de várias linhas com aspas triplas e f-string;
  • a formatar números: casas decimais, separador de milhar, porcentagem;
  • que, se o número mudar, você troca UMA variável e o relatório inteiro se atualiza.

Os números abaixo são do caso fictício (caso/dados/). Nos próximos dias você vai
aprender a calculá-los direto dos arquivos, em vez de digitá-los.

Como rodar (dentro da pasta dia-01-primeiro-script):
    python br/02_briefing_caso.py
"""

# ── Parâmetros ────────────────────────────────────────────────────────────
ANALISTA = "Analista"  #@param {type:"string"}

# ── 1. Os fatos do caso, cada um numa variável ────────────────────────────
# Texto (str) vai entre aspas; número inteiro (int) e decimal (float) vão sem aspas.
# Em Python o decimal usa PONTO: 54877.40, e não 54.877,40.
OPERACAO = "Vitrine Fantasma"
LOJA = "MegaOferta Brasil"
DOMINIO = "megaoferta-brasil.example"
SOLICITANTE = "Delegada Rita Assunção"
TOTAL_DENUNCIAS = 37
TOTAL_CIDADES = 11
CHAVES_PIX = 3
PREJUIZO_TOTAL = 54877.40          # soma dos valores das 37 denúncias, em reais
PRIMEIRA_DENUNCIA = "12/10/2025"
ULTIMA_DENUNCIA = "30/11/2025"
DIAS_DE_GOLPE = 49                 # 30/11 menos 12/10/2025 = 49 dias
PRODUTO_ISCA = "iPhone 15 128GB"
PRECO_ISCA = 1999.00
PRECO_REFERENCIA = 5000.00         # PARÂMETRO HIPOTÉTICO: troque pelo preço que você pesquisar hoje

# ── 2. Algumas contas com as variáveis ────────────────────────────────────
# O computador faz a conta; você só escreve a fórmula.
media_por_vitima = PREJUIZO_TOTAL / TOTAL_DENUNCIAS
denuncias_por_dia = TOTAL_DENUNCIAS / DIAS_DE_GOLPE
desconto = 1 - PRECO_ISCA / PRECO_REFERENCIA       # 0.60 significa 60%

# ── 3. O briefing ─────────────────────────────────────────────────────────
# Aspas triplas (""") deixam o texto ocupar várias linhas.
# Dentro das chaves, depois dos dois-pontos, vem o FORMATO:
#   :.2f  → 2 casas decimais        :,.2f → com separador de milhar
#   :.0%  → porcentagem sem casas   :>5   → alinhado à direita em 5 espaços
briefing = f"""
╔══════════════════════════════════════════════════════════════╗
  🕵️  NÚCLEO DE FONTES ABERTAS · BRIEFING
  Operação {OPERACAO}
╚══════════════════════════════════════════════════════════════╝

  Solicitante ........ {SOLICITANTE}
  Analista ........... {ANALISTA}

  🎯 Alvo
     Loja virtual "{LOJA}"
     Endereço na web: {DOMINIO}

  📨 Denúncias
     Total ........... {TOTAL_DENUNCIAS:>5}
     Cidades de MG ... {TOTAL_CIDADES:>5}
     Período ......... {PRIMEIRA_DENUNCIA} a {ULTIMA_DENUNCIA} ({DIAS_DE_GOLPE} dias)
     Ritmo ........... {denuncias_por_dia:.2f} denúncia(s) por dia

  💸 Dinheiro
     Prejuízo total .. R$ {PREJUIZO_TOTAL:,.2f}
     Média/vítima .... R$ {media_por_vitima:,.2f}
     Chaves PIX ...... {CHAVES_PIX} diferentes

  🪤 A isca
     {PRODUTO_ISCA} por R$ {PRECO_ISCA:,.2f}: {desconto:.0%} abaixo de R$ {PRECO_REFERENCIA:,.2f}
     (preço de referência hipotético; pesquise o real antes de usar num relatório).

  ❓ Perguntas para os próximos dias
     1. Os CPFs das denúncias são válidos?          (Dia 2)
     2. Existe algum perfil "kau.ofertas" público?   (Dia 3)
     3. O que as vítimas contaram nos depoimentos?   (Dia 4)
"""

print(briefing)

# ── 4. Atenção: o formato brasileiro de dinheiro ──────────────────────────
# O Python formata no padrão americano: 54,877.40. No Brasil escrevemos 54.877,40.
# Um truque simples (que você vai entender melhor no Dia 2): trocar os símbolos de lugar.
americano = f"{PREJUIZO_TOTAL:,.2f}"                                   # "54,877.40"
brasileiro = americano.replace(",", "X").replace(".", ",").replace("X", ".")
print(f"💡 Padrão americano: {americano}  →  padrão brasileiro: R$ {brasileiro}")
````

## ⚖️ Ética, LGPD e OPSEC

- **Briefing é documento sensível.** Mesmo um resumo com números pode revelar a linha de investigação.
  Não cole briefing de caso real em grupo de mensagens pessoal nem em ferramenta online pública.
- **Minimização (LGPD, art. 6º, III).** Repare que o briefing não lista nome nem CPF de nenhuma vítima:
  para o objetivo dele (dar a dimensão do caso), números agregados bastam.
- **Carimbo de tempo com fuso.** Horário sem fuso é uma fonte clássica de contestação. Guarde a hora
  em ISO 8601 com o deslocamento (`-03:00`).
- **Valores de referência precisam de fonte.** O preço de referência no script é hipotético. Num
  relatório real, todo número precisa dizer de onde veio e quando foi coletado.

# 🏆 Desafio S — Dia 1: Boletim de plantão

**Carta 43 · Rank S**

## A missão

Todo fim de turno, a Iara cola no grupo da equipe um **boletim curto** do caso. Ela cansou de digitar.
Escreva um script que monte esse boletim sozinho:

1. Parâmetros no topo: `ANALISTA`, `TOTAL_DENUNCIAS` (37), `PREJUIZO_TOTAL` (54877.40) e
   `PRIMEIRA_DENUNCIA` ("2025-10-12").
2. Primeira linha com **data e hora de Brasília** (`America/Sao_Paulo`, com plano B se o fuso faltar).
3. O prejuízo em **formato brasileiro**: `R$ 54.877,40` (e não `54,877.40`).
4. A **média por vítima**, também em formato brasileiro.
5. Quantos **dias se passaram** desde a primeira denúncia até hoje.
6. Imprima o boletim e, embaixo, quantos caracteres ele tem.

## Critérios de rank S

- [ ] Só biblioteca padrão, sem `input()`
- [ ] Hora com fuso (nada de `datetime.now()` "pelado")
- [ ] Dinheiro com ponto no milhar e vírgula no decimal
- [ ] Trocar um parâmetro atualiza o boletim inteiro

## Dica

Subtrair duas datas (`date`) dá um intervalo (`timedelta`). O atributo `.days` diz quantos dias ele tem.
Para o dinheiro: formate com `:,.2f` e depois troque `,` e `.` de lugar usando um caractere "coringa"
no meio do caminho (se trocar direto, um apaga o outro!).

````python
%cd -q {PASTA_NB}
# ✍️ Escreva sua solução aqui.
````

````python
#@title 🔒 Gabarito (dê dois cliques aqui para ver o código) { display-mode: "form" }
%cd -q {PASTA_NB}
"""
Gabarito do Desafio S do Dia 1 — Boletim de plantão.

Monta uma mensagem curta de plantão (do tipo que se cola no grupo da equipe) com:
carimbo de data/hora de Brasília, quantos dias se passaram desde a primeira denúncia,
o prejuízo em formato brasileiro (R$ 54.877,40) e a média por vítima.
Não usa internet.
"""

from datetime import date, datetime, timedelta, timezone

ANALISTA = "Analista"  #@param {type:"string"}
TOTAL_DENUNCIAS = 37  #@param {type:"integer"}
PREJUIZO_TOTAL = 54877.40  #@param {type:"number"}
PRIMEIRA_DENUNCIA = "2025-10-12"  #@param {type:"string"}

# Fuso de Brasília com plano B (Windows sem tzdata).
try:
    from zoneinfo import ZoneInfo
    FUSO = ZoneInfo("America/Sao_Paulo")
except Exception:
    FUSO = timezone(timedelta(hours=-3))

agora = datetime.now(FUSO)

# date.fromisoformat entende "AAAA-MM-DD". Subtrair duas datas dá um "timedelta";
# .days é o número de dias inteiros entre elas.
primeira = date.fromisoformat(PRIMEIRA_DENUNCIA)
dias_desde = (agora.date() - primeira).days

# Formato brasileiro: primeiro formatamos no padrão americano (vírgula no milhar,
# ponto no decimal) e depois trocamos os dois símbolos de lugar usando um "coringa".
americano = f"{PREJUIZO_TOTAL:,.2f}"
prejuizo_br = americano.replace(",", "§").replace(".", ",").replace("§", ".")
media = PREJUIZO_TOTAL / TOTAL_DENUNCIAS
media_br = f"{media:,.2f}".replace(",", "§").replace(".", ",").replace("§", ".")

boletim = f"""🚨 *BOLETIM DE PLANTÃO · NFA*
🕒 {agora:%d/%m/%Y %H:%M} (Brasília)
👤 {ANALISTA}

📨 Operação Vitrine Fantasma: {TOTAL_DENUNCIAS} denúncias
💸 Prejuízo: R$ {prejuizo_br} (média de R$ {media_br} por vítima)
⏳ Primeira denúncia há {dias_desde} dias ({primeira:%d/%m/%Y})
"""

print(boletim)
print(f"📏 Tamanho da mensagem: {len(boletim)} caracteres")
````
