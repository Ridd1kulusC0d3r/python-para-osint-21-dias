## 🧠 Conceito em 2 minutos

Existem três tipos de problema no código, e cada um tem a sua ferramenta:

| Problema | Exemplo | Quem acha |
|---|---|---|
| **Sintaxe**: o Python nem consegue ler | esqueceu o `:` no fim do `for` | `python -m py_compile`, `ast.parse`, o próprio editor |
| **Estilo**: roda, mas é difícil de ler | linhas enormes, nomes `x1`, `x2` | verificadores de estilo (PEP 8): `ruff`, `flake8` |
| **Lógica**: roda e dá o resultado **errado** | peso trocado no dígito verificador | **testes**: `doctest`, `unittest` |

O terceiro é o perigoso: nenhum verificador de sintaxe ou estilo vai notar. Por isso existem os
testes, que são exemplos com resposta conhecida:

```python
def dobro(x):
    """
    >>> dobro(21)
    42
    """
    return x * 2
```

`doctest.testmod()` roda os `>>>` da documentação e compara. O `unittest` organiza testes maiores
em classes, com `assertEqual`, `assertTrue` e companhia.

**IA** (ChatGPT e afins) ajuda nos três, mas **erra com confiança**. Trate a resposta como a de um
colega apressado: útil para dar ideias, nunca aceita sem testar.

### `original/nivel1_checar_sintaxe.py` — Verificador de sintaxe offline (ast + py_compile)

````python
%cd -q {PASTA_NB}/original
# ──────────────────────────────────────────────────────────────────────────
# Título: Verificador de sintaxe offline (ast + py_compile)
# NÃO há código do autor neste dia: o Day 21 do curso do cipher387 só recomenda
# ferramentas (verificadores de sintaxe e de estilo online, ChatGPT, Stack Overflow...).
# Este arquivo é uma RECONSTRUÇÃO da prática proposta, feita para este repositório:
# em vez de colar o código num site, o próprio Python verifica a sintaxe, sem internet
# (e sem mandar o seu código para ninguém).
# Uso:  python nivel1_checar_sintaxe.py [arquivo.py ...]   (padrão: codigo_com_erro.py)
# ──────────────────────────────────────────────────────────────────────────

import ast                 # "árvore de sintaxe abstrata": o Python lê o código sem executá-lo
import py_compile          # compila para bytecode (.pyc), como o Python faz antes de rodar
import sys
import tempfile
from pathlib import Path

PASTA = Path(__file__).resolve().parent if "__file__" in globals() else Path.cwd()
arquivos = [Path(a) for a in sys.argv[1:]] or [PASTA / "codigo_com_erro.py"]

com_erro = 0
for caminho in arquivos:
    codigo = caminho.read_text(encoding="utf-8")
    try:
        # ast.parse lê e entende a estrutura do código, mas NÃO executa nada.
        # Por isso é seguro usar até em código de origem duvidosa.
        arvore = ast.parse(codigo, filename=str(caminho))
    except SyntaxError as erro:
        com_erro += 1
        print(f"❌ {caminho.name}: erro de sintaxe na linha {erro.lineno}, coluna {erro.offset}")
        print(f"   Mensagem do Python: {erro.msg}")
        if erro.text:
            print(f"   {erro.text.rstrip()}")
            # Uma setinha embaixo do ponto exato do erro
            print("   " + " " * max((erro.offset or 1) - 1, 0) + "^")
        continue

    # Se passou no ast, contamos o que existe no arquivo (só para mostrar o que a árvore sabe).
    funcoes = [n.name for n in ast.walk(arvore) if isinstance(n, ast.FunctionDef)]
    importes = sorted({a.name.split(".")[0] for n in ast.walk(arvore) if isinstance(n, ast.Import) for a in n.names}
                      | {n.module.split(".")[0] for n in ast.walk(arvore) if isinstance(n, ast.ImportFrom) and n.module})
    # py_compile confirma compilando de verdade (num arquivo temporário, para não sujar a pasta).
    with tempfile.TemporaryDirectory() as tmp:
        py_compile.compile(str(caminho), cfile=str(Path(tmp) / "x.pyc"), doraise=True)
    print(f"✅ {caminho.name}: sintaxe ok · {len(codigo.splitlines())} linhas · "
          f"{len(funcoes)} funções · importa {', '.join(importes) or 'nada'}")

print(f"\n{len(arquivos)} arquivo(s) verificados, {com_erro} com erro de sintaxe.")
````

### `br/01_doctest_cpf.py` — doctest: exemplos dentro da documentação que viram testes automáticos.

````python
%cd -q {PASTA_NB}
"""
doctest: exemplos dentro da documentação que viram testes automáticos.

Carta 42 (rank A) — Jutsu do Revisor

O que você aprende:
  • a escrever exemplos no formato ">>> chamada / resultado esperado" dentro da docstring;
  • a rodar todos esses exemplos de uma vez com doctest.testmod();
  • a testar casos de borda (máscara, dígitos iguais, tamanho errado) da função de CPF do Dia 2;
  • que um teste vermelho é um AMIGO: ele avisa antes de o erro chegar ao relatório.

Como rodar (dentro da pasta dia-21-ferramentas-de-codigo):
    python br/01_doctest_cpf.py
    python -m doctest -v br/01_doctest_cpf.py     # mostra cada exemplo testado
"""

import doctest

# ── Parâmetros ────────────────────────────────────────────────────────────
MOSTRAR_CADA_TESTE = False  #@param {type:"boolean"}


def so_digitos(texto):
    """Devolve só os dígitos de um texto.

    >>> so_digitos("973.091.141-08")
    '97309114108'
    >>> so_digitos("sem números")
    ''
    """
    return "".join(c for c in texto if c.isdigit())


def cpf_valido(cpf):
    """Confere os dígitos verificadores de um CPF (a mesma regra do Dia 2, em forma de função).

    Com ou sem máscara, o resultado é o mesmo:

    >>> cpf_valido("973.091.141-08")
    True
    >>> cpf_valido("97309114108")
    True

    Um dígito trocado e a conta não fecha:

    >>> cpf_valido("973.091.141-09")
    False

    Casos de borda (são eles que pegam os bugs):

    >>> cpf_valido("111.111.111-11")      # todos iguais: a conta fecha, mas não vale
    False
    >>> cpf_valido("123.456.789")         # curto demais
    False
    >>> cpf_valido("")
    False

    Do caso: o CPF da DEN-007 não passa (alguém digitou errado na denúncia):

    >>> cpf_valido("084.520.599-47")
    False
    """
    d = [int(c) for c in so_digitos(cpf)]
    if len(d) != 11 or len(set(d)) == 1:
        return False
    for n in (9, 10):                                      # 1º e 2º dígito verificador
        soma = sum(d[i] * (n + 1 - i) for i in range(n))   # pesos 10..2 e depois 11..2
        dv = 0 if soma % 11 < 2 else 11 - soma % 11
        if dv != d[n]:
            return False
    return True


def formatar_cpf(cpf):
    """Põe a máscara 000.000.000-00. Levanta ValueError se não tiver 11 dígitos.

    >>> formatar_cpf("97309114108")
    '973.091.141-08'

    Exceções também são testáveis: o doctest compara a última linha do erro.

    >>> formatar_cpf("123")
    Traceback (most recent call last):
    ...
    ValueError: CPF precisa de 11 dígitos, veio 3
    """
    d = so_digitos(cpf)
    if len(d) != 11:
        raise ValueError(f"CPF precisa de 11 dígitos, veio {len(d)}")
    return f"{d[:3]}.{d[3:6]}.{d[6:9]}-{d[9:]}"


# ── Rodar todos os exemplos ───────────────────────────────────────────────
# testmod() procura ">>>" em todas as docstrings deste arquivo e confere os resultados.
resultado = doctest.testmod(verbose=MOSTRAR_CADA_TESTE)
print(f"🧪 doctest: {resultado.attempted} exemplos testados, {resultado.failed} falharam.")
if resultado.failed:
    raise SystemExit("❌ Algum exemplo da documentação não bate com o código. Leia o relatório acima.")
print("✅ A documentação e o código dizem a mesma coisa.")
print("💡 Experimente: troque '(n + 1 - i)' por '(n - i)' na função e rode de novo. Quem te avisa?")
````

### `br/02_unittest_regex.py` — unittest: uma bateria de testes para as expressões regulares do Dia 11.

````python
%cd -q {PASTA_NB}
"""
unittest: uma bateria de testes para as expressões regulares do Dia 11.

Carta 42 (rank A) — Jutsu do Revisor (parte 2)

O que você aprende:
  • a organizar testes em uma classe que herda de unittest.TestCase;
  • assertTrue, assertFalse, assertEqual e subTest (vários casos num teste só);
  • a testar o que DEVE casar e, tão importante quanto, o que NÃO deve casar;
  • a rodar os testes sem encerrar o Python (útil no Colab).

Como rodar (dentro da pasta dia-21-ferramentas-de-codigo):
    python br/02_unittest_regex.py
"""

import re
import unittest

# ── Parâmetros ────────────────────────────────────────────────────────────
DETALHADO = True  #@param {type:"boolean"}

# ── Os padrões (versões resumidas das do Dia 11, numa linha cada) ─────────
CPF = re.compile(r"(?<!\d)\d{3}\.?\d{3}\.?\d{3}-?\d{2}(?!\d)")
CNPJ = re.compile(r"(?<!\d)\d{2}\.?\d{3}\.?\d{3}/?\d{4}-?\d{2}(?!\d)")
CNPJ_ALFA = re.compile(r"(?<![A-Z0-9])[A-Z0-9]{2}\.?[A-Z0-9]{3}\.?[A-Z0-9]{3}/?[A-Z0-9]{4}-?\d{2}(?![A-Z0-9])")
TELEFONE = re.compile(r"(?<!\d)(?:\+?55[\s-]?)?\(?[1-9]\d\)?[\s-]?(?:9[\s-]?)?\d{4}[\s-]?\d{4}(?!\d)")
CEP = re.compile(r"(?<!\d)\d{5}-?\d{3}(?!\d)")
PIX_ALEATORIA = re.compile(r"\b[0-9a-f]{8}-[0-9a-f]{4}-[0-9a-f]{4}-[0-9a-f]{4}-[0-9a-f]{12}\b", re.IGNORECASE)
EMAIL = re.compile(r"[A-Za-z0-9._%+-]+@[A-Za-z0-9.-]+\.[A-Za-z]{2,}")


def casa_inteiro(padrao, texto):
    """True se o padrão casa com o texto INTEIRO (fullmatch), não só com um pedaço."""
    return padrao.fullmatch(texto) is not None


class TestaPadroesBR(unittest.TestCase):
    """Cada método que começa com 'test' é um teste. O unittest acha e roda todos."""

    def verificar(self, padrao, deve_casar, nao_deve_casar):
        # subTest: se um caso falhar, os outros continuam sendo testados e aparecem no relatório.
        for texto in deve_casar:
            with self.subTest(deveria_casar=texto):
                self.assertTrue(casa_inteiro(padrao, texto), f"deveria casar: {texto!r}")
        for texto in nao_deve_casar:
            with self.subTest(nao_deveria_casar=texto):
                self.assertFalse(casa_inteiro(padrao, texto), f"NÃO deveria casar: {texto!r}")

    def test_cpf(self):
        self.verificar(CPF, ["973.091.141-08", "97309114108"], ["973.091.141-0", "9730911410812", "abc.def.ghi-jk"])

    def test_cnpj(self):
        self.verificar(CNPJ, ["48.217.093/0001-30", "48217093000130"], ["48.217.093/0001", "4821709300013"])

    def test_cnpj_alfanumerico(self):
        # Formato novo da Receita (a partir de julho/2026): letras nas 12 primeiras posições.
        self.verificar(CNPJ_ALFA, ["12.ABC.345/01DE-35", "12ABC34501DE35"], ["12.ABC.345/01DE-3X", "12.abc.345/01de-35"])

    def test_telefone(self):
        self.verificar(TELEFONE, ["(31) 98888-0101", "+55 31 9 8888-0101", "31988880101", "3133334444"],
                       ["98888", "(01) 98888-0101"])          # DDD não começa com 0

    def test_cep(self):
        self.verificar(CEP, ["30130-010", "30130010"], ["3013-0010", "301300100"])

    def test_pix_aleatoria(self):
        self.verificar(PIX_ALEATORIA, ["f38b2ffc-80a4-df5a-51c9-bc701e7ea419", "F38B2FFC-80A4-DF5A-51C9-BC701E7EA419"],
                       ["f38b2ffc-80a4-df5a-51c9", "g38b2ffc-80a4-df5a-51c9-bc701e7ea419"])

    def test_email(self):
        self.verificar(EMAIL, ["financeiro@megaoferta-brasil.example", "kau.ofertas@example.com"],
                       ["kau.ofertas@", "@example.com", "kau ofertas@example.com"])

    def test_conversa_do_caso(self):
        """Teste de integração: numa linha real da conversa, quantas chaves PIX aparecem?"""
        linha = "Falta o PIX do frete, R$ 89,90 na chave f3f49249-dc28-ff90-a5ae-c7978306d03b"
        self.assertEqual(PIX_ALEATORIA.findall(linha), ["f3f49249-dc28-ff90-a5ae-c7978306d03b"])
        # E o valor "89,90" NÃO pode ser confundido com telefone nem CEP:
        self.assertEqual(TELEFONE.findall(linha), [])
        self.assertEqual(CEP.findall(linha), [])


# ── Rodar ─────────────────────────────────────────────────────────────────
# unittest.main() encerra o Python no fim (sys.exit). No Colab isso "derruba" a célula,
# então montamos a execução à mão: carregar os testes → rodar → olhar o resultado.
suite = unittest.defaultTestLoader.loadTestsFromTestCase(TestaPadroesBR)
resultado = unittest.TextTestRunner(verbosity=2 if DETALHADO else 1).run(suite)
print(f"\n🧪 {resultado.testsRun} testes · {len(resultado.failures)} falhas · {len(resultado.errors)} erros")
if not resultado.wasSuccessful():
    raise SystemExit("❌ Algum padrão quebrou. Leia o relatório acima: ele diz qual texto falhou.")
print("✅ Todos os padrões se comportam como o Dia 11 prometeu.")
````

### `br/03_dossie_final.py` — O grande final: o dossiê da Operação Vitrine Fantasma, gerado inteiro por código.

````python
%cd -q {PASTA_NB}
"""
O grande final: o dossiê da Operação Vitrine Fantasma, gerado inteiro por código.

Carta 42 (rank A) — Jutsu do Revisor (o dossiê)

O que você aprende (juntando os 21 dias):
  • RDAP + WHOIS (Dia 15): idade do domínio na 1ª vítima;
  • dump .sql no sqlite3 (Dia 8): quanto cada conta laranja recebeu;
  • CSV + padronização de cidades (Dias 7, 13, 16): onde estão as vítimas;
  • regex + dígito verificador (Dias 2, 11): os CNPJs citados pelo vendedor;
  • Wayback + datas (Dia 19): a linha do tempo e a Black Friday;
  • hashes SHA-256 (Dia 14) e relatório HTML (Dias 17 e 18): tudo auditável;
  • e autoverificações (Dia 21): o dossiê confere os próprios números antes de ser gravado.

100% offline e só com a biblioteca padrão.

Como rodar (dentro da pasta dia-21-ferramentas-de-codigo):
    python br/03_dossie_final.py
    (abra saidas/dossie.html no navegador; para PDF, Ctrl+P → Salvar como PDF)
"""

import calendar
import csv
import difflib
import hashlib
import html
import json
import re
import sqlite3
import unicodedata
from collections import Counter, defaultdict
from datetime import date, datetime, timedelta, timezone
from decimal import Decimal
from pathlib import Path

# ── Parâmetros ────────────────────────────────────────────────────────────
ANALISTA = "Analista NFA"  #@param {type:"string"}
DESTINATARIA = "Delegada Rita Assunção"  #@param {type:"string"}

PASTA_DIA = Path(__file__).resolve().parent.parent if "__file__" in globals() else Path.cwd()
PASTA_CASO = PASTA_DIA.parent / "caso" / "dados"
SAIDAS = PASTA_DIA / "saidas"

try:
    from zoneinfo import ZoneInfo
    FUSO = ZoneInfo("America/Sao_Paulo")
except Exception:
    FUSO = timezone(timedelta(hours=-3))

e = html.escape
MESES = ["janeiro", "fevereiro", "março", "abril", "maio", "junho", "julho",
         "agosto", "setembro", "outubro", "novembro", "dezembro"]
CIDADES_OFICIAIS = ["Belo Horizonte", "Betim", "Contagem", "Divinópolis", "Governador Valadares",
                    "Ipatinga", "Juiz de Fora", "Poços de Caldas", "Sete Lagoas", "Uberaba", "Uberlândia"]


# ── Ferramentas dos dias anteriores, em forma de função ───────────────────
def reais(v):                                                   # Dia 17
    return "R$ " + f"{v:,.2f}".replace(",", "§").replace(".", ",").replace("§", ".")


def br(d):                                                      # Dia 19
    return f"{d:%d/%m/%Y}"


def chave(t):                                                   # Dia 16
    s = "".join(c for c in unicodedata.normalize("NFKD", t) if not unicodedata.combining(c))
    return " ".join(s.casefold().split())


OFICIAIS = {chave(c): c for c in CIDADES_OFICIAIS}


def padronizar(cidade):                                         # Dia 7
    k = chave(cidade)
    if k in OFICIAIS:
        return OFICIAIS[k]
    p = difflib.get_close_matches(k, OFICIAIS, n=1, cutoff=0.8)
    return OFICIAIS[p[0]] if p else cidade


def cnpj_valido(cnpj):                                          # Dia 2
    """Aceita o formato numérico e o alfanumérico (valor = ord(c) - 48)."""
    c = re.sub(r"[^0-9A-Z]", "", cnpj.upper())
    if len(c) != 14 or not c[12:].isdigit() or len(set(c)) == 1:
        return False
    valores = [ord(x) - 48 for x in c]
    for n, pesos in ((12, [5, 4, 3, 2, 9, 8, 7, 6, 5, 4, 3, 2]), (13, [6, 5, 4, 3, 2, 9, 8, 7, 6, 5, 4, 3, 2])):
        resto = sum(v * p for v, p in zip(valores, pesos)) % 11
        if (0 if resto < 2 else 11 - resto) != valores[n]:
            return False
    return True


def sha256(caminho):                                            # Dia 14
    h = hashlib.sha256()
    with open(caminho, "rb") as f:
        for bloco in iter(lambda: f.read(65536), b""):
            h.update(bloco)
    return h.hexdigest()


def black_friday(ano):                                          # Dia 19
    quintas = [s[calendar.THURSDAY] for s in calendar.monthcalendar(ano, 11) if s[calendar.THURSDAY]]
    return date(ano, 11, quintas[3]) + timedelta(days=1)


# ── 1. Domínio (RDAP × WHOIS) ─────────────────────────────────────────────
ficha = json.loads((PASTA_CASO / "ficha_do_caso.json").read_text(encoding="utf-8"))
rdap = json.loads((PASTA_CASO / "rdap_megaoferta.json").read_text(encoding="utf-8"))
eventos_rdap = {ev["eventAction"]: ev["eventDate"] for ev in rdap["events"]}
criado = datetime.fromisoformat(eventos_rdap["registration"].replace("Z", "+00:00")).date()
titular_rdap = next(c[3] for ent in rdap["entities"] for c in ent["vcardArray"][1] if c[0] == "fn")
whois = (PASTA_CASO / "whois_megaoferta.txt").read_text(encoding="utf-8")
criado_whois = datetime.strptime(re.search(r"^created:\s+(\d{8})", whois, re.M).group(1), "%Y%m%d").date()
nameservers = [ns["ldhName"] for ns in rdap.get("nameservers", [])]

# ── 2. Denúncias ──────────────────────────────────────────────────────────
with open(PASTA_CASO / "denuncias.csv", encoding="utf-8") as f:
    denuncias = list(csv.DictReader(f, delimiter=";"))
total_denuncias = sum(Decimal(d["valor"]) for d in denuncias)
datas = sorted(date.fromisoformat(d["data"]) for d in denuncias)
cidades = Counter(padronizar(d["cidade"]) for d in denuncias)

# ── 3. Dump do painel (sqlite3 em memória) ────────────────────────────────
banco = sqlite3.connect(":memory:")
banco.executescript((PASTA_CASO / "painel_dump.sql").read_text(encoding="utf-8"))
laranjas = banco.execute("""
    SELECT c.titular, c.banco, c.chave_pix, COUNT(p.id), ROUND(SUM(p.valor), 2)
    FROM pedidos p JOIN contas_recebimento c ON c.id = p.conta_id
    GROUP BY c.id ORDER BY SUM(p.valor) DESC
""").fetchall()
total_dump = Decimal(str(banco.execute("SELECT ROUND(SUM(valor), 2) FROM pedidos").fetchone()[0]))
banco.close()

# ── 4. Conversa: CNPJs citados pelo vendedor ──────────────────────────────
conversa = (PASTA_CASO / "conversa_whatsapp.txt").read_text(encoding="utf-8")
cnpjs = sorted(set(re.findall(r"\b\d{2}\.\d{3}\.\d{3}/\d{4}-\d{2}\b", conversa)))

# ── 5. Wayback + linha do tempo ───────────────────────────────────────────
cab, *linhas_cdx = json.loads((PASTA_CASO / "wayback_cdx_megaoferta.json").read_text(encoding="utf-8"))
capturas = [dict(zip(cab, l)) for l in linhas_cdx]
bf = black_friday(2025)
perto_bf = sum(1 for d in datas if abs((d - bf).days) <= 3)

linha_do_tempo = [(criado, "Domínio registrado (RDAP e WHOIS concordam)" if criado == criado_whois
                   else "Domínio registrado (RDAP e WHOIS DISCORDAM)")]
linha_do_tempo += [(datas[0], "Primeira denúncia"), (bf, "Black Friday 2025"), (datas[-1], "Última denúncia")]
for c in capturas:
    dia = datetime.strptime(c["timestamp"], "%Y%m%d%H%M%S").replace(tzinfo=timezone.utc).astimezone(FUSO).date()
    linha_do_tempo.append((dia, f"Captura Wayback: HTTP {c['statuscode']}, {c['length']} bytes"))
linha_do_tempo.sort()

# ── 6. Evidências com hash ────────────────────────────────────────────────
evidencias = [(p.relative_to(PASTA_CASO).as_posix(), p.stat().st_size, sha256(p))
              for p in sorted(PASTA_CASO.rglob("*")) if p.is_file()]

# ── 7. Autoverificação: o dossiê confere a si mesmo ───────────────────────
# Se alguma fonte for alterada e os números deixarem de bater, o dossiê NÃO é gerado.
checagens = [
    ("Total de denúncias bate com a ficha do caso", len(denuncias) == ficha["total_denuncias"]),
    ("Soma das denúncias = soma dos pedidos no dump", total_denuncias == total_dump),
    ("RDAP e WHOIS concordam na data de criação", criado == criado_whois),
    ("CNPJ do rodapé é inválido (DV não fecha)", not cnpj_valido(ficha["cnpj_rodape"])),
    ("As 3 chaves do dump são as 3 chaves da ficha", {l[2] for l in laranjas} == set(ficha["chaves_pix"])),
    ("Todas as cidades foram padronizadas", all(c in CIDADES_OFICIAIS for c in cidades)),
]
print("🔎 Autoverificação do dossiê:")
for descricao, ok in checagens:
    print(f"   {'✅' if ok else '❌'} {descricao}")
if not all(ok for _, ok in checagens):
    raise SystemExit("❌ Os números não fecham. Corrija antes de entregar o dossiê.")

# ── 8. Escrever o HTML ────────────────────────────────────────────────────
agora = datetime.now(FUSO)
maior_cidade = max(cidades.values())


def tabela(cabecalho, linhas):
    th = "".join(f"<th>{e(c)}</th>" for c in cabecalho)
    trs = "".join("<tr>" + "".join(f"<td>{c}</td>" for c in linha) + "</tr>" for linha in linhas)
    return f"<table><tr>{th}</tr>{trs}</table>"


secao_laranjas = tabela(["Titular", "Instituição", "Chave PIX", "Pedidos", "Valor"],
                        [(e(t), e(b), f"<code>{e(c)}</code>", n, f"<span class='num'>{reais(Decimal(str(v)))}</span>")
                         for t, b, c, n, v in laranjas])
secao_cidades = "".join(
    f"<div class='barra'><span>{e(c)}</span><i style='width:{q / maior_cidade * 100:.0f}%'></i><b>{q}</b></div>"
    for c, q in cidades.most_common())
secao_cnpjs = tabela(["CNPJ citado na conversa", "Dígito verificador"],
                     [(f"<code>{e(c)}</code>", "✅ válido" if cnpj_valido(c) else "❌ inválido") for c in cnpjs])
secao_tempo = tabela(["Data", "D+", "Evento"], [(br(d), (d - criado).days, e(t)) for d, t in linha_do_tempo])
secao_hash = tabela(["Arquivo", "Bytes", "SHA-256"],
                    [(e(n), f"<span class='num'>{t}</span>", f"<code>{h}</code>") for n, t, h in evidencias])
secao_checagens = "".join(f"<li>{'✅' if ok else '❌'} {e(d)}</li>" for d, ok in checagens)

pagina = f"""<!doctype html>
<html lang="pt-BR"><head><meta charset="utf-8">
<meta name="viewport" content="width=device-width, initial-scale=1">
<title>Dossiê — Operação Vitrine Fantasma</title>
<style>
 :root {{ --azul: #1f5fa8; --texto: #1d2433; --cinza: #5b6475; --linha: #d5dbe5; }}
 body {{ font-family: system-ui, sans-serif; color: var(--texto); max-width: 980px; margin: 2rem auto; padding: 0 1rem; line-height: 1.45; }}
 header {{ border-bottom: 4px solid var(--azul); margin-bottom: 1.5rem; }}
 h1 {{ margin: .2rem 0; }} h2 {{ color: var(--azul); margin-top: 2rem; }}
 .meta {{ color: var(--cinza); font-size: .9rem; }}
 .cards {{ display: grid; grid-template-columns: repeat(auto-fit, minmax(170px, 1fr)); gap: .8rem; }}
 .card {{ background: #eef3fa; border-radius: 8px; padding: .7rem 1rem; }}
 .card b {{ display: block; font-size: 1.5rem; color: var(--azul); }}
 table {{ border-collapse: collapse; width: 100%; font-size: .88rem; margin: .5rem 0; }}
 th, td {{ border-bottom: 1px solid var(--linha); padding: .35rem .5rem; text-align: left; vertical-align: top; }}
 th {{ background: var(--azul); color: #fff; }} .num {{ white-space: nowrap; }}
 code {{ font-size: .78rem; word-break: break-all; }}
 .barra {{ display: grid; grid-template-columns: 170px 1fr 30px; align-items: center; gap: .5rem; margin: .2rem 0; }}
 .barra i {{ display: block; height: 16px; background: var(--azul); border-radius: 3px; }}
 .aviso {{ border-left: 4px solid #b3261e; background: #fbeeee; padding: .6rem 1rem; }}
 footer {{ margin-top: 2.5rem; color: var(--cinza); font-size: .8rem; border-top: 1px solid var(--linha); padding-top: .8rem; }}
 @media print {{ th {{ background: #ddd; color: #000; }} .barra i {{ background: #555; }} h2 {{ break-after: avoid; }} }}
</style></head>
<body>
<header>
 <div class="meta">NÚCLEO DE FONTES ABERTAS · uso restrito · dados fictícios de estudo</div>
 <h1>Dossiê — Operação Vitrine Fantasma</h1>
 <div class="meta">Para: {e(DESTINATARIA)} · Responsável: {e(ANALISTA)} ·
 Gerado em {agora.day} de {MESES[agora.month - 1]} de {agora.year}, {agora:%H:%M} (Brasília, UTC{agora:%z})</div>
</header>

<h2>1. Resumo executivo</h2>
<div class="cards">
 <div class="card"><b>{len(denuncias)}</b>vítimas em {len(cidades)} cidades de MG</div>
 <div class="card"><b>{reais(total_denuncias)}</b>prejuízo declarado</div>
 <div class="card"><b>{(datas[0] - criado).days} dias</b>de vida do domínio na 1ª vítima</div>
 <div class="card"><b>{len(laranjas)}</b>contas de recebimento</div>
 <div class="card"><b>{perto_bf / len(datas):.0%}</b>das denúncias em ±3 dias da Black Friday</div>
</div>
<p>A loja <code>{e(ficha['dominio'])}</code> foi registrada em {br(criado)} em nome de
<b>{e(titular_rdap)}</b>, anunciou eletrônicos muito abaixo do preço com pagamento exclusivo por PIX,
recebeu {len(denuncias)} pagamentos entre {br(datas[0])} e {br(datas[-1])}, passou a exibir uma página
mínima em 01/12/2025 e saiu do ar (HTTP 404) em 05/12/2025.</p>

<h2>2. Domínio</h2>
<ul>
 <li>Criação: RDAP {br(criado)} · WHOIS {br(criado_whois)} ({'as fontes concordam' if criado == criado_whois else 'DIVERGÊNCIA'})</li>
 <li>Titular declarado no registro: {e(titular_rdap)}</li>
 <li>Servidores DNS: {e(', '.join(nameservers))}</li>
 <li>CNPJ do rodapé {e(ficha['cnpj_rodape'])}: {'válido' if cnpj_valido(ficha['cnpj_rodape']) else '<b>dígito verificador inválido</b>'}</li>
</ul>

<h2>3. Para onde foi o dinheiro (dump do painel)</h2>
{secao_laranjas}
<p class="meta">Total no dump: {reais(total_dump)}, idêntico ao total declarado pelas vítimas.</p>
<p class="aviso">Hipótese a verificar: o titular "M. R. Brandão Comércio ME" tem o mesmo sobrenome que o
titular do domínio. Sobrenome igual <b>não</b> comprova vínculo.</p>

<h2>4. CNPJs citados pelo vendedor</h2>
{secao_cnpjs}

<h2>5. Onde estão as vítimas</h2>
{secao_cidades}

<h2>6. Linha do tempo</h2>
{secao_tempo}

<h2>7. Integridade das evidências</h2>
<p>SHA-256 de cada arquivo usado neste dossiê. Qualquer pessoa pode recalcular e confirmar que nada mudou.</p>
{secao_hash}

<h2>8. Autoverificação</h2>
<ul>{secao_checagens}</ul>

<h2>9. Limitações e próximos passos</h2>
<ul>
 <li>Os dados de registro do domínio e o nome dos titulares das contas são <b>declarados</b>: podem ser falsos ou de laranjas.</li>
 <li>As capturas da Wayback não cobrem todos os dias; há lacunas de até 30 dias sem prova direta do conteúdo do site.</li>
 <li>Sugere-se ofício às três instituições de pagamento e ao registrador do domínio.</li>
</ul>

<footer>Gerado por <code>br/03_dossie_final.py</code> (só biblioteca padrão do Python) a partir de
<code>caso/dados/</code>. Rodar de novo com os mesmos arquivos produz os mesmos números.</footer>
</body></html>
"""

SAIDAS.mkdir(exist_ok=True)
destino = SAIDAS / "dossie.html"
destino.write_text(pagina, encoding="utf-8")
hash_dossie = hashlib.sha256(pagina.encode("utf-8")).hexdigest()

print(f"\n📁 {destino.relative_to(PASTA_DIA)} gravado ({len(pagina):,} caracteres)".replace(",", "."))
print(f"🔏 SHA-256 do dossiê: {hash_dossie}")
print(f"   {len(denuncias)} vítimas · {reais(total_denuncias)} · {len(laranjas)} contas · domínio com "
      f"{(datas[0] - criado).days} dias na 1ª vítima")
print("📨 Pronto para entregar. Abra no navegador; para PDF, Ctrl+P → Salvar como PDF.")

if "__file__" not in globals():
    try:
        from IPython.display import HTML, display
        display(HTML(pagina))
    except ImportError:
        pass
````

> *"Ferramenta pronta todo mundo tem; quem sabe escrever a própria ferramenta enxerga o que os outros

O autor do curso original diz isso com todas as letras: o ChatGPT **erra muito**. Em 2026 os
assistentes de IA estão melhores, mas continuam inventando função que não existe, parâmetro que
mudou de nome e "fato" que ninguém escreveu. Em investigação, um erro desses vai parar num relatório
assinado por **você**. A regra deste repositório: **a IA sugere, você verifica, você assina.**

#### 1. Para que a IA é boa (e para que não é)

| Use para | Não use para |
|---|---|
| Explicar um trecho de código que você não entendeu | Decidir se uma pessoa é suspeita |
| Revisar seu script e apontar bugs prováveis | Ser a fonte de um fato do caso ("o CNPJ X é da empresa Y") |
| Sugerir casos de teste (doctest, unittest) | Gerar dados "de exemplo" que parecem reais e entram no relatório |
| Traduzir uma mensagem de erro em português claro | Resumir evidência sem você ler a evidência |
| Propor nomes melhores para variáveis e funções | Escrever a conclusão do dossiê |

#### 2. Um modelo de prompt para revisão de código

Copie, preencha e cole. Repare: **nenhum dado do caso vai junto**, só o código e dados fictícios.

```text
Você é um revisor de código Python experiente. Revise o script abaixo.

Contexto: script de estudo, só biblioteca padrão, Python 3.9+, roda no Windows e no Linux.
Objetivo do script: <uma frase>

Quero, nesta ordem:
1. Bugs que fariam o resultado sair ERRADO (não só feio). Para cada um: linha, por quê, correção.
2. Casos de borda que não estão tratados (arquivo vazio, acento, campo faltando, internet fora).
3. Riscos de segurança (injeção de HTML/SQL, path traversal, dados pessoais em log).
4. No máximo 3 sugestões de legibilidade.

Regras: não reescreva o script inteiro. Se não tiver certeza de algo, diga "não tenho certeza".
Cite a documentação oficial do Python quando falar de uma função da biblioteca padrão.

<cole o código aqui>
```

#### 3. Checagem de alucinação: 5 perguntas antes de aceitar uma resposta

1. **A função existe?** Abra a documentação oficial (`docs.python.org`) ou rode
   `python -c "import modulo; help(modulo.funcao)"`. Função inventada é a alucinação mais comum.
2. **O parâmetro tem esse nome nesta versão?** Bibliotecas mudam (lembra do `txt=` → `text=` do
   fpdf2 no Dia 17 e do `duckduckgo_search` → `ddgs` no Dia 9?).
3. **Eu consigo testar?** Escreva um doctest ou um unittest (scripts 01 e 02) **antes** de colar a
   sugestão no seu código. Se a IA disse que "isso corrige o bug", o teste tem que passar.
4. **A explicação bate com o que eu vejo rodando?** Rode com um exemplo pequeno e imprima o resultado.
5. **Isso é um FATO sobre o mundo?** (uma lei, uma data, quem é dono de um CNPJ, se uma API existe).
   Então a IA **não é fonte**. Procure a fonte primária: o texto da lei, a Receita, a documentação.

#### 4. Nunca cole dado pessoal de caso real em IA pública

- O que você digita num assistente público pode ser **guardado, revisado por pessoas ou usado para
  treinar** o modelo, conforme os termos do serviço. Para a LGPD, isso é um compartilhamento de dado
  pessoal com terceiro, e você precisa de base legal e de finalidade para fazer isso.
- Em caso real, troque tudo antes de pedir ajuda: nomes por `Pessoa A`, CPFs por `000.000.000-00`,
  chaves PIX por um UUID inventado, domínio por `exemplo.example`. **A IA precisa da estrutura do
  dado, não do dado.**
- Prefira ferramentas aprovadas pela sua instituição (contratos corporativos que não usam os dados
  para treino) ou modelos que rodam **localmente**, no seu computador.
- **OPSEC:** perguntar a um serviço externo "quem é o dono da chave PIX tal" pode revelar que existe
  uma investigação. Pergunte sobre **código**, nunca sobre **alvos**.

#### 5. O teste final

Se amanhã a Delegada Rita perguntar "de onde saiu este número?", você precisa responder com um
arquivo, uma linha de código e um hash. Nunca com "a IA disse".

## ⚖️ Ética, LGPD e OPSEC

- **Nunca cole dado de caso real numa IA pública** (nomes, CPFs, chaves PIX, prints de conversa).
  Anonimize: a IA precisa da *estrutura* do dado, não do dado. Detalhes em `br/04_ia_como_apoio.md`.
- **IA não é fonte.** Se o relatório diz "o CNPJ X pertence a Y", a fonte tem que ser a Receita (ou
  outra primária), nunca "o assistente disse".
- **Perguntar também revela.** Buscar no Stack Overflow ou numa IA "como achar o dono da chave PIX
  f38b…" expõe o alvo e a investigação. Pergunte sobre **código**, com exemplos inventados.
- **Testes protegem pessoas.** Um bug no validador de CPF pode descartar uma denúncia verdadeira ou
  "confirmar" um documento falso. Testar é parte da responsabilidade de quem assina o relatório.
- **O dossiê vai circular.** Ele traz titulares das contas e o nome declarado no domínio: tudo é
  **declaração**, e o documento diz isso na seção de limitações. Hipótese não vira fato por estar
  bem formatada.

# 🏆 Desafio S — Dia 21: O guardião do repositório

**Carta 63 · Rank S**

## A missão

O curso acabou, mas o repositório vai continuar recebendo contribuições: novos exercícios, correções,
dias extras. A Iara quer um **guardião**: um script que leia todos os exercícios BR e aponte quem
quebrou as regras do [`CONTRIBUTING.md`](https://github.com/Ridd1kulusC0d3r/python-para-osint-21-dias/blob/main/CONTRIBUTING.md), **sem executar nada**.

Escreva um script que, para cada `dia-*/br/*.py` do repositório:

1. Confira a **sintaxe** com `ast.parse` (e não pare se um arquivo tiver erro).
2. Exija **docstring** no topo com a palavra "Carta" (o Dia 0 é exceção: não tem cartas).
3. Acuse qualquer chamada a **`input()`**.
4. Acuse **imports fora da biblioteca padrão**, exceto os que estão dentro de um
   `try/except ImportError` (como o `IPython` e o `google.colab`, que são opcionais).
5. Acuse **`open()` em modo texto** e `read_text()`/`write_text()` **sem `encoding=`**.
6. Imprima um placar por dia e grave `saidas/revisao_codigo.csv`.

## Critérios de rank S

- [ ] Só biblioteca padrão; nenhum script analisado é executado ou importado
- [ ] `open(arquivo, "rb")` **não** é acusado (modo binário não tem encoding)
- [ ] Um `import requests` solto é acusado; um `from IPython.display import HTML` dentro de `try/except ImportError` não é
- [ ] Funciona no Python 3.9 (dica: `sys.stdlib_module_names` só existe a partir do 3.10)
- [ ] Testado com um arquivo "ruim" feito por você (cada regra acusada pelo menos uma vez)

## Dica

`ast.walk(arvore)` visita todos os nós. Os que interessam: `ast.Call` (chamadas de função; olhe
`no.func` e `no.keywords`), `ast.Import`, `ast.ImportFrom` e `ast.Try` (com `no.handlers`).
`ast.get_docstring(arvore)` devolve a docstring do módulo.

````python
%cd -q {PASTA_NB}
# ✍️ Escreva sua solução aqui.
````

````python
#@title 🔒 Gabarito (dê dois cliques aqui para ver o código) { display-mode: "form" }
%cd -q {PASTA_NB}
"""
Gabarito do Desafio S do Dia 21 — O revisor automático do curso.

Lê (SEM executar) todos os br/*.py do repositório com o módulo ast e confere as regras do
CONTRIBUTING.md:
  1. a sintaxe está ok;
  2. tem docstring no topo, com a palavra "Carta" (menos no Dia 0, que não tem cartas);
  3. não usa input();
  4. só importa biblioteca padrão (imports dentro de try/except ImportError contam como opcionais);
  5. todo open()/read_text()/write_text() em modo TEXTO informa encoding= (armadilha do Windows).
Grava saidas/revisao_codigo.csv e imprime um placar por dia.
"""

import ast
import csv
import sys
from collections import defaultdict
from pathlib import Path

SO_PROBLEMAS = False  #@param {type:"boolean"}

PASTA_DIA = Path(__file__).resolve().parent if "__file__" in globals() else Path.cwd()
RAIZ = PASTA_DIA.parent
SAIDAS = PASTA_DIA / "saidas"

# sys.stdlib_module_names existe a partir do Python 3.10. No 3.9, usamos uma lista mínima.
STDLIB = set(getattr(sys, "stdlib_module_names", ())) or {
    "argparse", "ast", "bisect", "calendar", "collections", "csv", "datetime", "decimal", "difflib",
    "doctest", "glob", "hashlib", "html", "http", "itertools", "json", "locale", "math", "os", "pathlib",
    "platform", "py_compile", "random", "re", "shutil", "socket", "sqlite3", "string", "subprocess",
    "sys", "tempfile", "threading", "time", "unicodedata", "unittest", "urllib", "uuid", "zoneinfo",
}


def imports_opcionais(arvore):
    """Nomes importados dentro de um bloco try cujo except trata ImportError (ou Exception)."""
    opcionais = set()
    for no in ast.walk(arvore):
        if isinstance(no, ast.Try):
            tratados = {getattr(h.type, "id", None) for h in no.handlers}
            if tratados & {"ImportError", "ModuleNotFoundError", "Exception", None}:
                for filho in no.body:
                    for sub in ast.walk(filho):
                        if isinstance(sub, ast.Import):
                            opcionais |= {a.name.split(".")[0] for a in sub.names}
                        elif isinstance(sub, ast.ImportFrom) and sub.module:
                            opcionais.add(sub.module.split(".")[0])
    return opcionais


def modo_do_open(chamada):
    """Devolve o modo de um open(...) se for um texto fixo; senão, 'r' (o padrão)."""
    if len(chamada.args) >= 2 and isinstance(chamada.args[1], ast.Constant):
        return str(chamada.args[1].value)
    for kw in chamada.keywords:
        if kw.arg == "mode" and isinstance(kw.value, ast.Constant):
            return str(kw.value.value)
    return "r"


def revisar(caminho):
    problemas = []
    sem_cartas = caminho.parts[-3].startswith("dia-00")     # o Dia 0 não tem cartas
    codigo = caminho.read_text(encoding="utf-8")
    try:
        arvore = ast.parse(codigo, filename=str(caminho))
    except SyntaxError as erro:
        return [f"sintaxe: linha {erro.lineno}: {erro.msg}"]

    doc = ast.get_docstring(arvore)
    if not doc:
        problemas.append("sem docstring no topo")
    elif "Carta" not in doc and not sem_cartas:
        problemas.append("docstring sem a linha 'Carta'")

    opcionais = imports_opcionais(arvore)
    for no in ast.walk(arvore):
        # 3. input()
        if isinstance(no, ast.Call) and isinstance(no.func, ast.Name) and no.func.id == "input":
            problemas.append(f"linha {no.lineno}: usa input()")
        # 4. imports fora da stdlib
        nomes = []
        if isinstance(no, ast.Import):
            nomes = [a.name.split(".")[0] for a in no.names]
        elif isinstance(no, ast.ImportFrom) and no.module and no.level == 0:
            nomes = [no.module.split(".")[0]]
        for nome in nomes:
            if nome not in STDLIB and nome not in opcionais:
                problemas.append(f"linha {no.lineno}: importa '{nome}' (não é stdlib)")
        # 5. encoding
        if isinstance(no, ast.Call):
            tem_encoding = any(kw.arg == "encoding" for kw in no.keywords)
            if isinstance(no.func, ast.Name) and no.func.id == "open":
                if "b" not in modo_do_open(no) and not tem_encoding:
                    problemas.append(f"linha {no.lineno}: open() em modo texto sem encoding=")
            elif isinstance(no.func, ast.Attribute) and no.func.attr in ("read_text", "write_text"):
                if not tem_encoding:
                    problemas.append(f"linha {no.lineno}: {no.func.attr}() sem encoding=")
    return problemas


# ── Varrer o repositório ──────────────────────────────────────────────────
linhas = []
placar = defaultdict(lambda: [0, 0])                 # dia → [arquivos, arquivos com problema]
for arquivo in sorted(RAIZ.glob("dia-*/br/*.py")):
    dia = arquivo.parts[-3]
    problemas = revisar(arquivo)
    placar[dia][0] += 1
    placar[dia][1] += bool(problemas)
    linhas.append({"dia": dia, "arquivo": arquivo.name, "ok": "sim" if not problemas else "não",
                   "problemas": " | ".join(problemas)})
    if problemas or not SO_PROBLEMAS:
        print(f"{'✅' if not problemas else '⚠️ '} {dia}/br/{arquivo.name}")
        for p in problemas:
            print(f"      • {p}")

print("\n📋 Placar por dia (arquivos · com apontamento):")
for dia, (total, ruins) in sorted(placar.items()):
    print(f"   {dia:<40} {total:>2} · {ruins}")
total = sum(v[0] for v in placar.values())
ruins = sum(v[1] for v in placar.values())
print(f"\n🧮 {total} scripts revisados, {ruins} com apontamentos. Nada foi executado: só lido com ast.")

SAIDAS.mkdir(exist_ok=True)
with open(SAIDAS / "revisao_codigo.csv", "w", newline="", encoding="utf-8-sig") as f:
    escritor = csv.DictWriter(f, fieldnames=["dia", "arquivo", "ok", "problemas"], delimiter=";")
    escritor.writeheader()
    escritor.writerows(linhas)
print("💾 saidas/revisao_codigo.csv gravado.")
````
