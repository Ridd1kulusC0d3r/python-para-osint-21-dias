## 🧠 Conceito em 2 minutos

Quatro peças bastam para começar a automatizar:

| Peça | Para quê | Exemplo |
|---|---|---|
| **Variável** | guardar um valor com um nome | `cpf = "973.091.141-08"` |
| **Condição** (`if` / `elif` / `else`) | fazer uma coisa **ou** outra | `if len(digitos) != 11:` |
| **Lista** | guardar vários valores em ordem | `cpfs = ["973...", "188..."]` |
| **Laço** (`for`) | repetir algo para cada item | `for cpf in cpfs:` |

O que fica "dentro" de um `if` ou de um `for` é marcado pelo **recuo** (4 espaços). Errou o recuo,
o Python reclama ou, pior, faz outra coisa.

**Dígito verificador** é um número calculado a partir dos outros para pegar erros de digitação.
No CPF: multiplique os 9 primeiros dígitos pelos pesos 10, 9, …, 2, some, pegue o resto da divisão
por 11; se o resto for 0 ou 1 o DV é 0, senão é 11 − resto. Repita com 10 dígitos e pesos 11, …, 2.
No CNPJ a ideia é a mesma, com pesos `5 4 3 2 9 8 7 6 5 4 3 2` (e um `6` na frente para o 2º DV). No
**CNPJ alfanumérico**, cada caractere vale `ord(caractere) − 48`: os algarismos continuam valendo eles
mesmos, e as letras vão de A = 17 até Z = 42.

### `br/01_variaveis_br.py` — Variáveis à brasileira: tipos de dado e dinheiro no formato R$ 1.999,00.

````python
%cd -q {PASTA_NB}
"""
Variáveis à brasileira: tipos de dado e dinheiro no formato R$ 1.999,00.

Carta 23 (rank B) — Jutsu da Triagem (parte 1)

O que você aprende:
  • os quatro tipos básicos: texto (str), inteiro (int), decimal (float) e verdadeiro/falso (bool);
  • a descobrir o tipo de qualquer valor com type();
  • a converter de um tipo para outro (int("37"), str(37), float("149.90"));
  • por que "37" + 1 dá erro e como resolver;
  • a escrever uma função que mostra dinheiro no padrão brasileiro.

Como rodar (dentro da pasta dia-02-sintaxe-basica):
    python br/01_variaveis_br.py
"""

# ── Parâmetros (no Colab viram um formulário) ─────────────────────────────
VALOR_PIX = 1999.0  #@param {type:"number"}
VALOR_DIGITADO = "1.999,00"  #@param {type:"string"}

# ── 1. Os quatro tipos básicos ────────────────────────────────────────────
# Em Python você não precisa declarar o tipo: ele é deduzido pelo valor.
vitima = "Mariana Lima Carvalho"     # str: texto, sempre entre aspas
parcelas = 1                         # int: número inteiro
valor = 1899.00                      # float: número com casas decimais (ponto, não vírgula!)
pagou_no_pix = True                  # bool: True (verdadeiro) ou False (falso)

print("🧪 Os quatro tipos básicos")
# type() diz o tipo; .__name__ mostra só o nome curto (str, int...).
for nome, conteudo in [("vitima", vitima), ("parcelas", parcelas),
                       ("valor", valor), ("pagou_no_pix", pagou_no_pix)]:
    print(f"   {nome:<13} = {conteudo!r:<26} → {type(conteudo).__name__}")
# O !r mostra o valor "como o Python vê" (texto aparece com aspas).

# ── 2. Converter tipos ────────────────────────────────────────────────────
# Dados que chegam de formulários, planilhas e sites quase sempre chegam como TEXTO.
print("\n🔄 Convertendo tipos")
total_texto = "37"
print(f"   '37' é {type(total_texto).__name__}; int('37') + 1 = {int(total_texto) + 1}")
print(f"   str(37) + ' denúncias' = {str(37) + ' denúncias'!r}")
print(f"   float('149.90') = {float('149.90')}")

# Tentar somar texto com número dá TypeError. O try/except mostra o erro sem parar o script.
try:
    print(total_texto + 1)
except TypeError as erro:
    print(f"   ❌ '37' + 1 → TypeError: {erro}")

# ── 3. Nomes de variáveis ─────────────────────────────────────────────────
# Pode: letras, números (menos no começo) e _. Use nomes que expliquem o conteúdo.
# Pode acento? Pode (valor_médio funciona), mas o costume é evitar em nomes de variáveis.
cidade_da_vitima = "Divinópolis"     # bom: diz o que é
x = "Divinópolis"                    # ruim: daqui a uma semana você não lembra o que é x
print(f"\n🏷️  Nome claro: cidade_da_vitima = {cidade_da_vitima!r}  (melhor que x = {x!r})")


# ── 4. Dinheiro no formato brasileiro ─────────────────────────────────────
# Uma FUNÇÃO é um bloco de código com nome que você pode reaproveitar.
# def cria a função; o que vem entre parênteses são as entradas; return devolve o resultado.
def reais(numero):
    """Transforma 1999.0 em 'R$ 1.999,00'."""
    americano = f"{numero:,.2f}"                      # '1,999.00' (milhar com vírgula, decimal com ponto)
    # Trocamos os símbolos usando um "coringa" (§) para um não apagar o outro.
    brasileiro = americano.replace(",", "§").replace(".", ",").replace("§", ".")
    return f"R$ {brasileiro}"


def ler_reais(texto):
    """O caminho inverso: transforma 'R$ 1.999,00' (ou '1.999,00') em 1999.0."""
    limpo = texto.replace("R$", "").strip()           # tira o símbolo e os espaços das pontas
    limpo = limpo.replace(".", "").replace(",", ".")  # '1.999,00' → '1999.00'
    return float(limpo)


print("\n💸 Dinheiro à brasileira")
for v in (VALOR_PIX, 149.9, 54877.4, 0.5, 1234567.891):
    print(f"   {v!r:>12} → {reais(v)}")

convertido = ler_reais(VALOR_DIGITADO)
print(f"\n   Texto digitado {VALOR_DIGITADO!r} → número {convertido} → de volta: {reais(convertido)}")

# ── 5. Um cuidado com float ───────────────────────────────────────────────
# O computador guarda decimais em binário, e alguns valores não "fecham" exatamente.
print(f"\n⚠️  0.1 + 0.2 = {0.1 + 0.2}  (por isso, para somar dinheiro de verdade, o Dia 7 usa Decimal)")
````

### `br/02_valida_cpf.py` — Validar CPF: o algoritmo dos dígitos verificadores, passo a passo.

````python
%cd -q {PASTA_NB}
"""
Validar CPF: o algoritmo dos dígitos verificadores, passo a passo.

Carta 23 (rank B) — Jutsu da Triagem (parte 2)

O que você aprende:
  • a percorrer os caracteres de um texto com for e guardar só os dígitos;
  • a usar if/elif/else para rejeitar entradas com tamanho errado ou dígitos todos iguais;
  • a calcular os dois dígitos verificadores (DV) do CPF com laços e o resto da divisão (%);
  • que "CPF válido" quer dizer só "a conta fecha" — não que a pessoa existe.

Como rodar (dentro da pasta dia-02-sintaxe-basica):
    python br/02_valida_cpf.py

O CPF padrão é fictício (de caso/dados/denuncias.csv).
"""

# ── Parâmetros ────────────────────────────────────────────────────────────
CPF = "973.091.141-08"  #@param {type:"string"}
MOSTRAR_CONTAS = True  #@param {type:"boolean"}

# ── Como funciona (leia antes do código) ──────────────────────────────────
# Um CPF tem 11 dígitos: 9 "de base" + 2 verificadores (DV).
#   1º DV: multiplique os 9 primeiros dígitos pelos pesos 10, 9, 8, ..., 2 e some tudo.
#          Pegue o resto da divisão da soma por 11.  Resto < 2 → DV = 0.  Senão → DV = 11 − resto.
#   2º DV: a mesma coisa com os 10 primeiros dígitos (base + 1º DV) e pesos 11, 10, ..., 2.
# Se os dois DV calculados forem iguais aos dois últimos dígitos do CPF, a conta fecha.

# ── Passo 1: ficar só com os dígitos ──────────────────────────────────────
# Percorremos cada caractere; isdigit() é True para '0'...'9'. Pontos e traço ficam de fora.
digitos = []
for caractere in CPF:
    if caractere.isdigit():
        digitos.append(int(caractere))      # guardamos como número para poder multiplicar

print(f"🔢 Entrada: {CPF!r} → dígitos: {digitos}")

# ── Passo 2: checagens básicas ────────────────────────────────────────────
# if / elif / else: só UM dos blocos roda, o primeiro cuja condição for verdadeira.
valido = True
motivo = ""
if len(digitos) != 11:
    valido = False
    motivo = f"tem {len(digitos)} dígitos (precisa de 11)"
elif len(set(digitos)) == 1:
    # set() remove repetidos. Se sobrar 1 elemento, todos são iguais (ex.: 111.111.111-11).
    # Esses CPFs "fecham a conta", mas a Receita não os considera válidos.
    valido = False
    motivo = "todos os dígitos são iguais"
else:
    # ── Passo 3: calcular os dois dígitos verificadores ───────────────────
    calculados = []
    for posicao in (9, 10):                 # 1ª volta: usa 9 dígitos; 2ª volta: usa 10
        soma = 0
        peso = posicao + 1                  # 10 na 1ª volta, 11 na 2ª
        base = digitos[:posicao]            # fatia: os primeiros 9 (ou 10) dígitos
        for digito in base:
            if MOSTRAR_CONTAS:
                print(f"      {digito} × {peso:>2} = {digito * peso:>3}")
            soma += digito * peso           # += é "some ao que já tinha"
            peso -= 1                       # o peso diminui 1 a cada dígito
        resto = soma % 11                   # % é o resto da divisão
        dv = 0 if resto < 2 else 11 - resto
        print(f"   ➕ soma = {soma}; {soma} % 11 = {resto} → DV {len(calculados) + 1} = {dv}")
        calculados.append(dv)

    informados = digitos[9:]                # os dois últimos dígitos do CPF
    print(f"   🎯 DV calculados: {calculados}  ·  DV informados: {informados}")
    if calculados != informados:
        valido = False
        motivo = "os dígitos verificadores não batem"

# ── Passo 4: o veredito ───────────────────────────────────────────────────
if valido:
    print(f"\n✅ {CPF} é VÁLIDO (a conta dos dígitos verificadores fecha).")
    print("   Lembre-se: válido ≠ existente. Só a Receita confirma se o CPF existe e de quem é.")
else:
    print(f"\n❌ {CPF} é INVÁLIDO: {motivo}.")
    print("   Pode ser erro de digitação da vítima, ou um número inventado. Anote e confira o original.")
````

### `br/03_triagem_cpfs.py` — Triagem de CPFs: validar uma lista inteira de denúncias e contar o resultado.

````python
%cd -q {PASTA_NB}
"""
Triagem de CPFs: validar uma lista inteira de denúncias e contar o resultado.

Carta 23 (rank B) — Jutsu da Triagem (parte 3, aplicação no caso)

O que você aprende:
  • a guardar vários registros numa lista de listas ([id, cpf]);
  • a transformar o validador do script 02 numa função reutilizável;
  • a percorrer a lista com for, decidir com if e ir contando com variáveis;
  • a montar uma pequena tabela com ✅ / ❌ e um resumo no final.

Os 10 CPFs abaixo foram copiados de caso/dados/denuncias.csv (fictícios).
No Dia 7 você vai aprender a ler o CSV inteiro direto do arquivo.

Como rodar (dentro da pasta dia-02-sintaxe-basica):
    python br/03_triagem_cpfs.py
"""

# ── Parâmetros ────────────────────────────────────────────────────────────
MOSTRAR_SO_INVALIDOS = False  #@param {type:"boolean"}

# ── A lista de denúncias (id, CPF informado pela vítima) ──────────────────
# Cada item da lista é outra lista com dois elementos. [0] é o id, [1] é o CPF.
DENUNCIAS = [
    ["DEN-005", "978.753.477-18"],
    ["DEN-006", "267.147.513-62"],
    ["DEN-007", "084.520.599-47"],
    ["DEN-008", "628.360.161-83"],
    ["DEN-009", "177.992.298-10"],
    ["DEN-010", "073.702.838-60"],
    ["DEN-011", "477.647.258-97"],
    ["DEN-012", "205.258.983-01"],
    ["DEN-013", "621.052.185-17"],
    ["DEN-014", "304.118.955-06"],
]


def cpf_valido(cpf):
    """Devolve True se a conta dos dígitos verificadores do CPF fecha."""
    digitos = [int(c) for c in cpf if c.isdigit()]   # mesma ideia do script 02, numa linha só
    if len(digitos) != 11 or len(set(digitos)) == 1:
        return False
    for posicao in (9, 10):
        soma = 0
        for i in range(posicao):
            # Peso começa em posicao+1 (10 ou 11) e diminui 1 a cada dígito.
            soma += digitos[i] * (posicao + 1 - i)
        resto = soma % 11
        dv = 0 if resto < 2 else 11 - resto
        if dv != digitos[posicao]:
            return False                              # na primeira falha, já sabemos a resposta
    return True


# ── A triagem ─────────────────────────────────────────────────────────────
validos = 0                  # contadores começam em zero
invalidos = 0
para_conferir = []           # aqui guardamos os ids que precisam de atenção

print(f"{'Denúncia':<9}  {'CPF':<15}  Resultado")
print("─" * 40)
for denuncia in DENUNCIAS:
    identificador = denuncia[0]
    cpf = denuncia[1]
    if cpf_valido(cpf):
        validos += 1
        if not MOSTRAR_SO_INVALIDOS:
            print(f"{identificador:<9}  {cpf:<15}  ✅ válido")
    else:
        invalidos += 1
        para_conferir.append(identificador)
        print(f"{identificador:<9}  {cpf:<15}  ❌ inválido")

# ── Resumo ────────────────────────────────────────────────────────────────
total = len(DENUNCIAS)
print("─" * 40)
print(f"📊 {total} CPFs analisados: {validos} válidos e {invalidos} inválidos "
      f"({invalidos / total:.0%} da amostra).")

if para_conferir:
    print(f"🔍 Pedir à equipe para conferir os documentos originais de: {', '.join(para_conferir)}")
    print("   CPF inválido numa denúncia quase sempre é erro de digitação, não fraude da vítima.")
````

### `br/04_valida_cnpj.py` — Validar CNPJ — numérico e o novo CNPJ alfanumérico (desde julho de 2026).

````python
%cd -q {PASTA_NB}
"""
Validar CNPJ — numérico e o novo CNPJ alfanumérico (desde julho de 2026).

Carta 23 (rank B) — Jutsu da Triagem (parte 4)

O que você aprende:
  • a calcular os dois dígitos verificadores do CNPJ com pesos e módulo 11;
  • a usar ord() para transformar um caractere em número (valor = ord(caractere) − 48);
  • por que a MESMA conta serve para o CNPJ antigo (só números) e para o novo (com letras);
  • a primeira pista do caso: o CNPJ do rodapé da loja não passa na validação.

Contexto: a Receita Federal (IN RFB nº 2.229/2024) passou a emitir, a partir de julho de 2026,
CNPJs com letras e números nas 12 primeiras posições. O primeiro foi gerado em 31/07/2026,
para uma filial do Banco do Brasil: 00.000.000/E08G-12. Os dois últimos (os DV) continuam
numéricos, e os CNPJs antigos não mudam.

Como rodar (dentro da pasta dia-02-sintaxe-basica):
    python br/04_valida_cnpj.py
"""

# ── Parâmetros ────────────────────────────────────────────────────────────
CNPJ_RODAPE = "48.217.093/0001-30"  #@param {type:"string"}

# Pesos oficiais. O 1º DV usa 12 pesos; o 2º DV usa 13 (um 6 na frente).
PESOS_DV1 = [5, 4, 3, 2, 9, 8, 7, 6, 5, 4, 3, 2]
PESOS_DV2 = [6, 5, 4, 3, 2, 9, 8, 7, 6, 5, 4, 3, 2]


def valor(caractere):
    """Valor de um caractere na conta: ord() dá o código na tabela ASCII.
    '0' tem código 48, então '0'→0 ... '9'→9. 'A' tem 65, então 'A'→17 ... 'Z'→42."""
    return ord(caractere) - 48


def limpar(cnpj):
    """Tira pontos, barra, traço e espaços; põe as letras em maiúsculas."""
    limpo = ""
    for c in cnpj.upper():
        if c.isalnum():              # isalnum = letra ou número
            limpo += c
    return limpo


def digito(base, pesos):
    """Um dígito verificador: soma de valor × peso, resto por 11."""
    soma = 0
    for caractere, peso in zip(base, pesos):     # zip anda nas duas listas ao mesmo tempo
        soma += valor(caractere) * peso
    resto = soma % 11
    return "0" if resto < 2 else str(11 - resto)


def cnpj_valido(cnpj):
    """True se o CNPJ (numérico ou alfanumérico) tem os DV corretos."""
    c = limpar(cnpj)
    if len(c) != 14:
        return False
    if not c[12:].isdigit():                      # os dois DV são sempre números
        return False
    if len(set(c)) == 1:                          # 00000000000000, 11111111111111...
        return False
    dv1 = digito(c[:12], PESOS_DV1)
    dv2 = digito(c[:12] + dv1, PESOS_DV2)
    return c[12:] == dv1 + dv2


def explicar(cnpj):
    """Mostra a conta inteira, passo a passo."""
    c = limpar(cnpj)
    print(f"\n🔎 {cnpj}  →  limpo: {c}")
    if len(c) != 14:
        print(f"   ❌ tem {len(c)} caracteres (precisa de 14)")
        return
    base = c[:12]
    for numero, pesos in ((1, PESOS_DV1), (2, PESOS_DV2)):
        partes = [f"{ch}({valor(ch)})×{p}" for ch, p in zip(base, pesos)]
        soma = sum(valor(ch) * p for ch, p in zip(base, pesos))
        dv = digito(base, pesos)
        print(f"   DV{numero}: " + " + ".join(partes))
        print(f"        soma = {soma}; {soma} % 11 = {soma % 11} → DV{numero} = {dv}")
        base += dv                                # o 2º DV usa a base + o 1º DV
    esperado = base[12:]
    if c[12:] == esperado:
        print(f"   ✅ VÁLIDO: os DV informados ({c[12:]}) batem com os calculados ({esperado}).")
    else:
        print(f"   ❌ INVÁLIDO: informado {c[12:]}, mas a conta dá {esperado}.")


# ── 1. Exemplos ───────────────────────────────────────────────────────────
print("🏦 CNPJ numérico (Banco do Brasil, dado público):")
explicar("00.000.000/0001-91")

print("\n🔤 CNPJ alfanumérico (o primeiro emitido pela Receita: filial do Banco do Brasil, 31/07/2026):")
explicar("00.000.000/E08G-12")

# ── 2. A pista do caso ────────────────────────────────────────────────────
print("\n" + "═" * 60)
print("🕵️  Caso Vitrine Fantasma: o CNPJ impresso no rodapé da loja")
explicar(CNPJ_RODAPE)
if not cnpj_valido(CNPJ_RODAPE):
    print("\n🚩 PISTA: o CNPJ do rodapé não passa nem na conta dos dígitos verificadores.")
    print("   Empresa de verdade não erra o próprio CNPJ. Ele foi inventado (ou copiado errado).")
    print("   Isso já entra no relatório. No Dia 6, você vai ver que nem vale gastar uma consulta com ele.")

# ── 3. Resumo rápido ──────────────────────────────────────────────────────
print("\n📋 Resumo")
for teste in ("00.000.000/0001-91", "00.000.000/E08G-12", "12.ABC.345/01DE-35", CNPJ_RODAPE, "11.111.111/1111-11", "12.345"):
    print(f"   {teste:<22} {'✅ válido' if cnpj_valido(teste) else '❌ inválido'}")
````

## ⚖️ Ética, LGPD e OPSEC

- **CPF é dado pessoal** (LGPD, art. 5º, I). Em investigação, trate só os necessários para a finalidade e
  nunca publique listas de CPFs, nem "de exemplo". Todos os deste repositório são fictícios.
- **Válido não é existente.** O validador só confere a matemática. Quem diz se o CPF existe e de quem
  é, é a Receita Federal, pelos canais legais.
- **Não "conserte" documentos para consultá-los.** Se um CPF ou CNPJ está inválido, o número com o DV
  "corrigido" provavelmente é **de outra pessoa ou empresa**. Consultá-lo é tratar dado de um terceiro
  sem motivo. O certo é voltar à fonte (a vítima, o documento original).
- **CPF inválido numa denúncia não é indício contra a vítima.** É, quase sempre, erro de digitação.
- **CNPJ inválido no site de uma loja é indício**, sim: empresas reais copiam o CNPJ do cartão do CNPJ.

## A missão

## Critérios de rank S

## Dica

E `range(10, 1, -1)` gera os pesos do CPF: 10, 9, 8, ..., 2.

````python
%cd -q {PASTA_NB}
# ✍️ Escreva sua solução aqui.
````

````python
#@title 🔒 Gabarito (dê dois cliques aqui para ver o código) { display-mode: "form" }
%cd -q {PASTA_NB}
"""
Gabarito do Desafio S do Dia 2 — Classificador de documentos.

Recebe uma lista misturada de "números de documento" (como os que aparecem soltos numa
conversa de WhatsApp) e, para cada um, diz:
  • o tipo provável (CPF, CNPJ numérico, CNPJ alfanumérico ou desconhecido);
  • se é válido;
  • se for inválido, quais seriam os dígitos verificadores corretos.
Só biblioteca padrão; não usa internet.
"""

DOCUMENTOS = [
    "48.217.093/0001-30",     # CNPJ do rodapé da loja (caso)
    "31.560.427/0001-71",     # "CNPJ do sócio" citado na conversa (caso)
    "019.018.791-35",         # CPF que um cliente mandou "pra nota" (caso)
    "084.520.599-47",         # CPF da denúncia DEN-007 (caso)
    "00.000.000/0001-91",     # Banco do Brasil (público)
    "00.000.000/E08G-12",     # 1º CNPJ alfanumérico emitido (filial do Banco do Brasil, público)
    "(31) 98888-0101",        # telefone com 11 dígitos: NÃO pode ser confundido com CPF
]

PESOS_CNPJ_1 = [5, 4, 3, 2, 9, 8, 7, 6, 5, 4, 3, 2]
PESOS_CNPJ_2 = [6] + PESOS_CNPJ_1


def mod11(soma):
    resto = soma % 11
    return 0 if resto < 2 else 11 - resto


def dv_cpf(nove):
    """Recebe os 9 primeiros dígitos (texto) e devolve os 2 DV corretos (texto)."""
    numeros = [int(c) for c in nove]
    d1 = mod11(sum(n * p for n, p in zip(numeros, range(10, 1, -1))))
    d2 = mod11(sum(n * p for n, p in zip(numeros + [d1], range(11, 1, -1))))
    return f"{d1}{d2}"


def dv_cnpj(doze):
    """Recebe as 12 primeiras posições (texto, pode ter letras) e devolve os 2 DV."""
    valores = [ord(c) - 48 for c in doze]
    d1 = mod11(sum(v * p for v, p in zip(valores, PESOS_CNPJ_1)))
    d2 = mod11(sum(v * p for v, p in zip(valores + [d1], PESOS_CNPJ_2)))
    return f"{d1}{d2}"


def classificar(texto):
    """Devolve (tipo, válido?, DV corretos) para um texto qualquer."""
    limpo = "".join(c for c in texto.upper() if c.isalnum())
    tem_letra = any(c.isalpha() for c in limpo)

    # Telefone também pode ter 11 dígitos! A pista é o formato: parênteses no DDD ou +55.
    if "(" in texto or texto.strip().startswith("+"):
        return "telefone", False, "—"

    if len(limpo) == 11 and not tem_letra and "/" not in texto:
        if len(set(limpo)) == 1:
            return "CPF", False, "—"
        certo = dv_cpf(limpo[:9])
        return "CPF", limpo[9:] == certo, certo

    if len(limpo) == 14 and limpo[12:].isdigit():
        tipo = "CNPJ alfanumérico" if tem_letra else "CNPJ numérico"
        if len(set(limpo)) == 1:
            return tipo, False, "—"
        certo = dv_cnpj(limpo[:12])
        return tipo, limpo[12:] == certo, certo

    return "desconhecido", False, "—"


print(f"{'Documento':<22} {'Tipo':<18} {'Resultado':<12} DV corretos")
print("─" * 66)
contagem = {}
for doc in DOCUMENTOS:
    tipo, ok, certo = classificar(doc)
    contagem[tipo] = contagem.get(tipo, 0) + 1          # dicionário como contador
    if tipo in ("desconhecido", "telefone"):
        resultado = "⚪ ignorado"
    else:
        resultado = "✅ válido" if ok else "❌ inválido"
    print(f"{doc:<22} {tipo:<18} {resultado:<12} {certo if not ok else ''}")

print("─" * 66)
print("📊 " + " · ".join(f"{tipo}: {n}" for tipo, n in contagem.items()))
print("\n🕵️  Os dois CNPJs que o vendedor usou são inválidos: nenhuma empresa pode ter esses números.")
print("   Repare: 'DV corretos' mostra qual seria o final certo. NÃO use isso para 'consertar'")
print("   um documento e consultá-lo: o número corrigido provavelmente é de outra pessoa ou empresa.")
````

## 🔎 Inovação OSINT do dia

**CNPJ alfanumérico em vigor.** Desde **julho de 2026**, a Receita Federal (Instrução Normativa RFB
nº 2.229/2024) pode emitir CNPJs com **letras e números** nas 12 primeiras posições, mantendo os 2
dígitos verificadores numéricos. O primeiro saiu em **31/07/2026**, para uma filial do Banco do
Brasil: `00.000.000/E08G-12` (o `br/04` usa esse número público como exemplo e ele passa na conta).
CNPJs antigos continuam valendo, e números só com algarismos ainda podem ser emitidos. Para OSINT isso quebra muita coisa:
regex que só aceita `\d{14}`, planilhas que guardam CNPJ como número (e somem com as letras e os
zeros à esquerda) e validadores antigos. A boa notícia: a conta do DV é a mesma, trocando `int(c)`
por `ord(c) - 48`. Biblioteca simples: nenhuma, só a função nativa `ord()`. Confira sempre a
documentação oficial da Receita antes de usar o validador em produção.
